## Czego się uczymy i po co?

Macierz $P$ mówi, co się dzieje w **jednym** kroku. Tu nauczymy się odpowiadać na pytanie: *jakie jest prawdopodobieństwo, że za $n$ kroków będę w stanie $j$?* Odpowiedź okaże się piękna: wystarczy policzyć **$n$-tą potęgę macierzy** $P^n$. Poznamy też pojęcie **rozkładu niezmienniczego** — pierwszego kroku do analizy długoterminowego zachowania łańcucha.

### Ważne rozróżnienie notacyjne

- $n$ = **liczba kroków** (czas),
- $N$ = **liczba stanów** (rozmiar macierzy).

To dwie różne litery i dwie różne rzeczy! Dalej: $p_{ij}^{(n)}$ to prawdopodobieństwo przejścia w $n$ krokach — **górny indeks w nawiasie to nie potęga liczby** $p_{ij}$, tylko etykieta „$n$ kroków". Natomiast cała macierz spełnia $P^{(n)} = P^n$ (prawdziwa potęga macierzowa) — i to właśnie udowodnimy.

## 📘 Prawdopodobieństwo przejścia w $n$ krokach (§1.2)

$$p_{ij}^{(n)} := P(X_{k+n} = j \mid X_k = i) = P(X_n = j \mid X_0 = i)$$

(druga równość dzięki jednorodności — nie ma znaczenia, kiedy startujemy).

### Wzór Chapmana–Kołmogorowa (1.20)

$$p_{ij}^{(n)} = \sum_{h \in E} p_{ih}^{(n-1)}\, p_{hj}$$

**Intuicja:** żeby dojść z $i$ do $j$ w $n$ krokach, trzeba po $n-1$ krokach być w **jakimś** stanie $h$, a potem w jednym kroku przejść $h \to j$. Sumujemy po wszystkich możliwych „przystankach" $h$. Dowód w podręczniku to rozpisanie prawdopodobieństwa warunkowego i użycie własności Markowa.

### Konsekwencja: $P^{(n)} = P^n$

Wzór C–K mówi dokładnie, że macierz $P^{(n)} = (p_{ij}^{(n)})$ spełnia $P^{(n)} = P^{(n-1)} P$ (mnożenie macierzowe!). Przez indukcję:

$$P^{(2)} = P\cdot P = P^2, \qquad P^{(3)} = P^2 \cdot P = P^3, \qquad \dots \qquad P^{(n)} = P^n.$$

Z Esercizio 1.1 (notatnik 01) wiemy, że $P^n$ jest nadal macierzą stochastyczną. ✔

## 📘 Rozkład łańcucha w chwili $n$ (§1.3)

Rozkład $X_n$ opisują liczby $w_i = P(X_n = i)$, zebrane w **wektor wierszowy** $w = (w_1, \dots, w_N)$, $w_i \ge 0$, $\sum_i w_i = 1$.

Jeśli $X_0$ ma rozkład (wł. *legge*, *distribuzione*) $u = (u_1,\dots,u_N)$, to z prawdopodobieństwa całkowitego (wzór (1.21)):

$$w_i = P(X_n = i) = \sum_{h\in E} P(X_n = i \mid X_0 = h)\,P(X_0 = h) = \sum_{h\in E} p_{hi}^{(n)}\, u_h,$$

czyli macierzowo (wzór (1.22)):

$$w = u\,P^n$$

**Uwaga podręcznika:** $P^n$ mnożymy **z lewej** przez wektor wierszowy $u$. (Drobna uwaga redakcyjna: w tekście źródłowym wektory $w$, $u$ mają symbol transpozycji $(\cdot)^T$, ale są używane jako wektory wierszowe — stosujemy konsekwentnie zapis wierszowy, zgodny z faktycznym rachunkiem $w = uP^n$.)

Przypadek szczególny: start pewny ze stanu $k$ ($u_h = \delta_{hk}$, delta Kroneckera: 1 gdy $h=k$, 0 w przeciwnym razie) daje $w_i = p_{ki}^{(n)}$ — czyli **wiersz $k$ macierzy $P^n$** to rozkład po $n$ krokach przy starcie z $k$.

## 📘 Definicja 1.3 (rozkład niezmienniczy / distribuzione invariante)

Rozkład $v$ na $E$ jest **niezmienniczy** dla łańcucha, jeżeli

$$v = vP.$$

Jeśli rozkład początkowy $u$ jest niezmienniczy, to (wzór (1.25)) $w = uP^n = u$ dla każdego $n$: łańcuch „wygląda tak samo" w każdej chwili. **Uwaga terminologiczna:** to jeszcze NIE jest „rozkład stacjonarny" w sensie Abundo — o subtelnej różnicy opowiada notatnik 06.

## Przykład liczbowy krok po kroku (🧪 własny)

Weźmy urządzenie z Przykładu 1.2 (notatnik 01): $P = \begin{pmatrix} 0.8 & 0.2 \\ 0.1 & 0.9 \end{pmatrix}$ (stan 0 = zepsute, 1 = sprawne).

**Krok 1.** Policzmy $P^2$ ręcznie — element $(0,1)$, czyli $P(\text{sprawne za 2 dni}\mid\text{zepsute dziś})$:

$$p_{01}^{(2)} = p_{00}p_{01} + p_{01}p_{11} = 0.8\cdot 0.2 + 0.2\cdot 0.9 = 0.16 + 0.18 = 0.34.$$

Dwie drogi: „zepsute→zepsute→sprawne" (0.16) i „zepsute→sprawne→sprawne" (0.18). Chapman–Kołmogorow to dokładnie suma po drogach!

**Krok 2.** Rozkład po $n$ krokach ze startu $u = (1, 0)$: $w = uP^n$ = pierwszy wiersz $P^n$.

**Krok 3.** Rozkład niezmienniczy: rozwiązujemy $v = vP$ z $v_1 + v_2 = 1$:

$$v_1 = 0.8 v_1 + 0.1 v_2 \;\Longrightarrow\; 0.2 v_1 = 0.1 v_2 \;\Longrightarrow\; v_2 = 2 v_1 \;\Longrightarrow\; v = \left(\tfrac13, \tfrac23\right).$$

```python
import numpy as np
from fractions import Fraction

P = np.array([[0.8, 0.2],
              [0.1, 0.9]])

# Krok 1: P^2 przez prawdziwe mnożenie macierzowe
P2 = np.linalg.matrix_power(P, 2)
print("P^2 =\n", P2)
print("p_01^(2) =", P2[0, 1], " (ręcznie: 0.34)")
```

```python
# Krok 2: ewolucja rozkładu w = u P^n
u = np.array([1.0, 0.0])            # start: na pewno zepsute
for n in [1, 2, 5, 10, 50]:
    w = u @ np.linalg.matrix_power(P, n)
    print(f"n = {n:3d}:  w = {np.round(w, 5)}")
# widać zbieżność do (1/3, 2/3)
```

```python
# Krok 3: rozkład niezmienniczy dokładnie, na ułamkach
Pf = [[Fraction(4,5), Fraction(1,5)],
      [Fraction(1,10), Fraction(9,10)]]
# v = vP i v1+v2=1  =>  v1(1-p00) = v2 p10
# 0.2 v1 = 0.1 v2  => v2 = 2 v1, v1 = 1/3
v = (Fraction(1,3), Fraction(2,3))
vP = (v[0]*Pf[0][0] + v[1]*Pf[1][0], v[0]*Pf[0][1] + v[1]*Pf[1][1])
print("v  =", v)
print("vP =", vP, " -> niezmienniczy:", vP == v)
```

## Typowe błędy

1. **$p_{ij}^{(n)} \ne (p_{ij})^n$!** Lewa strona to element macierzy $P^n$; prawa to $n$-ta potęga pojedynczej liczby. W kodzie: `matrix_power(P, n)[i, j]`, nigdy `P[i, j]**n`.
2. Mylenie $n$ (kroki) z $N$ (liczba stanów).
3. Mnożenie $P^n u$ zamiast $u P^n$ — przy wierszowo-stochastycznej konwencji rozkład zawsze z lewej.
4. Zakładanie, że $w = uP^n$ zawsze zbiega do rozkładu niezmienniczego — to wymaga dodatkowych założeń (notatniki 06–07)!

## Podsumowanie

- Chapman–Kołmogorow: $p_{ij}^{(n)} = \sum_h p_{ih}^{(n-1)} p_{hj}$, czyli $P^{(n)} = P^n$.
- Rozkład w chwili $n$: $w = u P^n$ (wektory wierszowe, mnożenie z lewej).
- Wiersz $k$ macierzy $P^n$ = rozkład po $n$ krokach przy starcie z $k$.
- Rozkład niezmienniczy: $v = vP$ — gdy od niego startujemy, rozkład nie zmienia się nigdy.

## Pytania kontrolne

1. Dla $P$ z przykładu policz ręcznie $p_{11}^{(2)}$ (sprawne → sprawne w 2 krokach).
2. Czy suma $\sum_j p_{ij}^{(n)} = 1$ dla każdego $i$ i $n$? Dlaczego?
3. Niech $u = (\tfrac13, \tfrac23)$ dla naszego $P$. Co będzie rozkładem $X_{1000}$?

*(Odpowiedzi poniżej.)*

## Odpowiedzi

1. $p_{11}^{(2)} = p_{10}p_{01} + p_{11}p_{11} = 0.1\cdot 0.2 + 0.9 \cdot 0.9 = 0.02 + 0.81 = 0.83$.
2. Tak — $P^n$ jest macierzą stochastyczną (iloczyn macierzy stochastycznych, Esercizio 1.1), a wiersze macierzy stochastycznej sumują się do 1.
3. Dokładnie $u = (\tfrac13, \tfrac23)$ — bo $u$ jest rozkładem niezmienniczym, więc $uP^{1000} = u$.
