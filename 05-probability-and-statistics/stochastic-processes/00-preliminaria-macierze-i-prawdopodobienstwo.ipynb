## 1. Prawdopodobieństwo warunkowe (probabilità condizionale)

### Intuicja

Prawdopodobieństwo warunkowe $P(A\mid B)$ czytamy: *„prawdopodobieństwo, że zajdzie $A$, jeżeli już wiemy, że zaszło $B$"*. Wiedza o $B$ „zawęża świat" — liczymy szanse $A$ tylko wśród tych scenariuszy, w których $B$ się wydarzyło.

### Definicja

Jeżeli $P(B) > 0$, to

$$P(A \mid B) = \frac{P(A \cap B)}{P(B)}.$$

Symbol po symbolu:
- $P(A\mid B)$ — prawdopodobieństwo $A$ **pod warunkiem** $B$ (kreska $\mid$ = „pod warunkiem", wł. *dato che*),
- $P(A\cap B)$ — prawdopodobieństwo, że zajdą **oba** zdarzenia naraz,
- dzielimy przez $P(B)$, żeby „przeskalować": w nowym, zawężonym świecie zdarzenie $B$ ma mieć prawdopodobieństwo 1.

Z definicji od razu wynika **wzór łańcuchowy** (używany w podręczniku przy wzorach (1.2)–(1.5)):

$$P(A \cap B) = P(A \mid B)\, P(B).$$

### Mały przykład (🧪 własny)

Rzucamy zwykłą kostką. $B$ = „wypadła liczba parzysta" $=\{2,4,6\}$, $A$ = „wypadła szóstka" $=\{6\}$.

$$P(A\mid B) = \frac{P(\{6\})}{P(\{2,4,6\})} = \frac{1/6}{3/6} = \frac{1}{3}.$$

Wiedza, że wynik jest parzysty, podniosła szansę szóstki z $1/6$ do $1/3$.

```python
from fractions import Fraction

# Przykład z kostką: sprawdzamy definicję P(A|B) = P(A i B) / P(B)
omega = [1, 2, 3, 4, 5, 6]          # wszystkie wyniki, każdy z prawd. 1/6
B = {2, 4, 6}                        # warunek: liczba parzysta
A = {6}                              # zdarzenie: szóstka

P = lambda zb: Fraction(len(zb), len(omega))
P_A_given_B = P(A & B) / P(B)
print("P(A|B) =", P_A_given_B)       # oczekujemy 1/3
```

## 2. Wektory i macierze — wiersze i kolumny

### Co to jest macierz?

Macierz $N \times M$ to prostokątna tabela liczb: $N$ **wierszy** (wł. *righe*) i $M$ **kolumn** (wł. *colonne*). Element $a_{ij}$ stoi w wierszu $i$ i kolumnie $j$ — **pierwszy indeks to zawsze wiersz, drugi to kolumna**.

$$A = \begin{pmatrix} a_{11} & a_{12} \\ a_{21} & a_{22} \end{pmatrix}$$

- **Wektor wierszowy** (wł. *vettore riga*) to macierz $1\times N$: $u = (u_1, u_2, \dots, u_N)$.
- **Wektor kolumnowy** (wł. *vettore colonna*) to macierz $N\times 1$.

### Dlaczego to ważne dla łańcuchów Markowa?

W podręczniku Abundo macierz przejścia $P$ jest **wierszowo-stochastyczna**: element $p_{ij}$ to prawdopodobieństwo przejścia **ze stanu $i$ (wiersz) do stanu $j$ (kolumna)**, a każdy **wiersz sumuje się do 1**. Rozkłady prawdopodobieństwa (np. rozkład początkowy) zapisujemy jako **wektory wierszowe** i mnożymy je przez $P$ **z lewej strony**: $\mu_{n+1} = \mu_n P$. Pomylenie wierszy z kolumnami to najczęstszy błąd początkujących!

## 3. Mnożenie macierzy „wiersz razy kolumna" (prodotto riga per colonna)

### Definicja

Jeżeli $A$ ma wymiar $N\times K$, a $B$ wymiar $K\times M$, to iloczyn $C = AB$ ma wymiar $N \times M$ i

$$c_{ij} = \sum_{h=1}^{K} a_{ih}\, b_{hj}.$$

Słowami: element $c_{ij}$ to **iloczyn skalarny wiersza $i$ macierzy $A$ z kolumną $j$ macierzy $B$** — bierzemy wiersz, bierzemy kolumnę, mnożymy po współrzędnych, sumujemy.

### Przykład krok po kroku (🧪 własny)

$$\begin{pmatrix} 1 & 2 \\ 3 & 4 \end{pmatrix}\begin{pmatrix} 5 & 6 \\ 7 & 8 \end{pmatrix} = \begin{pmatrix} 1\cdot 5 + 2\cdot 7 & 1\cdot 6 + 2\cdot 8 \\ 3\cdot 5 + 4\cdot 7 & 3\cdot 6 + 4\cdot 8 \end{pmatrix} = \begin{pmatrix} 19 & 22 \\ 43 & 50 \end{pmatrix}$$

### Wektor wierszowy razy macierz

Dla wektora wierszowego $u=(u_1,\dots,u_N)$ i macierzy $P$ wynik $uP$ jest znowu wektorem wierszowym:

$$(uP)_j = \sum_{h=1}^{N} u_h\, p_{hj}.$$

Tak właśnie „przesuwa się" rozkład prawdopodobieństwa o jeden krok łańcucha Markowa.

### Potęga macierzy to NIE potęgowanie elementów!

$P^2 = P\cdot P$ (mnożenie macierzowe), a **nie** macierz z elementami $p_{ij}^2$. W NumPy: `np.linalg.matrix_power(P, n)` — nigdy `P**n` dla tablic `ndarray` (to podnosi do potęgi każdy element osobno). Uwaga: w Scilabie jest odwrotnie — tam `P^n` **jest** potęgą macierzową.

```python
import numpy as np

A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("A @ B (mnożenie macierzowe):")
print(A @ B)

print("\nmatrix_power(A, 2) = A @ A:")
print(np.linalg.matrix_power(A, 2))

print("\nUWAGA — A**2 w NumPy to potęgowanie PO ELEMENTACH (nie tego chcemy!):")
print(A**2)
```

## 4. Wartości własne i wektory własne (autovalori e autovettori)

### Intuicja

Mnożenie przez macierz $A$ zwykle „obraca i rozciąga" wektory. **Wektor własny** to wyjątkowy kierunek, którego macierz nie obraca — tylko go skaluje. Współczynnik skalowania to **wartość własna** $\lambda$.

### Definicje — uwaga na stronę mnożenia!

- **Prawy wektor własny** (kolumnowy): $u \ne 0$ taki, że $A u = \lambda u$. Tu $u$ jest **wektorem kolumnowym** i stoi **po prawej** stronie $A$.
- **Lewy wektor własny** (wierszowy): $v \ne 0$ taki, że $v A = \lambda v$. Tu $v$ jest **wektorem wierszowym** i stoi **po lewej**.

Dla łańcuchów Markowa (z wierszowo-stochastyczną $P$):
- rozkład stacjonarny/niezmienniczy $\pi$ spełnia $\pi P = \pi$, więc $\pi$ to **lewy** wektor własny $P$ dla $\lambda = 1$;
- wektor kolumnowy samych jedynek $\mathbf{1}$ spełnia $P\mathbf{1} = \mathbf{1}$ (bo wiersze sumują się do 1), więc $\mathbf{1}$ to **prawy** wektor własny dla $\lambda = 1$.

Lewe wektory własne $P$ to prawe wektory własne macierzy transponowanej $P^T$ — tego triku używa się w obliczeniach numerycznych.

### Jak się je znajduje?

Wartości własne to pierwiastki **wielomianu charakterystycznego** $\det(A - \lambda I) = 0$. Dla macierzy $2\times 2$ to równanie kwadratowe. Potem dla każdego $\lambda$ rozwiązujemy układ $(A-\lambda I)u = 0$.

```python
import numpy as np

# 🧪 Przykład własny: macierz stochastyczna (wiersze sumują się do 1)
P = np.array([[0.9, 0.1],
              [0.5, 0.5]])

# prawe wektory własne: P u = lambda u  (kolumny macierzy 'wektory')
wartosci, wektory = np.linalg.eig(P)
print("wartości własne P:", wartosci)        # zawsze jest 1 wśród nich!

# wektor jedynek to prawy wektor własny dla lambda=1:
print("P @ [1,1] =", P @ np.ones(2))

# lewy wektor własny dla lambda=1 -> prawy wektor własny P^T:
w, v = np.linalg.eig(P.T)
k = np.argmin(abs(w - 1))                     # znajdź lambda ~ 1
pi = np.real(v[:, k])
pi = pi / pi.sum()                            # normalizacja: suma = 1
print("lewy wektor własny (rozkład stacjonarny) pi =", pi)
print("sprawdzenie pi @ P =", pi @ P)         # ma wyjść to samo pi
```

## Typowe błędy

1. **Mylenie wierszy z kolumnami w $P$** — u Abundo $p_{ij}$ = z $i$ do $j$, wiersze sumują się do 1. W niektórych innych książkach (zwłaszcza fizycznych) macierz jest kolumnowo-stochastyczna — zawsze sprawdź konwencję!
2. **`P**n` zamiast `np.linalg.matrix_power(P, n)`** w NumPy — potęgowanie elementów zamiast macierzy.
3. **Mylenie lewych i prawych wektorów własnych** — $\pi P = \pi$ (lewy, wierszowy) to co innego niż $Pu = u$ (prawy, kolumnowy).
4. **Zapominanie o normalizacji** — wektor własny jest wyznaczony z dokładnością do stałej; żeby był rozkładem prawdopodobieństwa, trzeba go podzielić przez sumę współrzędnych (i sprawdzić, że współrzędne są $\ge 0$).

## Podsumowanie

- $P(A\mid B) = P(A\cap B)/P(B)$; stąd $P(A\cap B) = P(A\mid B)P(B)$.
- W macierzy pierwszy indeks = wiersz, drugi = kolumna; $(AB)_{ij} = \sum_h a_{ih}b_{hj}$.
- Rozkłady to wektory wierszowe, ewoluują przez $\mu_{n+1} = \mu_n P$.
- $\pi P = \pi$: rozkład niezmienniczy to lewy wektor własny $P$ dla $\lambda=1$.

## Odpowiedzi

1. Zdarzenia: $B$ = „co najmniej jeden orzeł" ma 3 z 4 równo prawdopodobnych wyników (OO, OR, RO), $A\cap B$ = „dwa orły" = 1 wynik. Zatem $P(A\mid B) = \frac{1/4}{3/4} = \frac{1}{3}$.
2. $(uP)_1 = 0.3\cdot 0.9 + 0.7\cdot 0.5 = 0.27 + 0.35 = 0.62$; $(uP)_2 = 0.3\cdot 0.1 + 0.7\cdot 0.5 = 0.03+0.35 = 0.38$. Wynik: $uP = (0.62,\ 0.38)$ — nadal suma 1!
3. Bo $P\mathbf{1} = \mathbf{1}$: licząc $i$-tą współrzędną $\sum_j p_{ij}\cdot 1 = 1$ (wiersze sumują się do 1). Skoro istnieje niezerowy wektor $u=\mathbf{1}$ z $Pu = 1\cdot u$, to $1$ jest wartością własną.
