## Czego się uczymy i po co?

Poznajemy najważniejszy obiekt całego kursu: **łańcuch Markowa** (wł. *Catena di Markov*, w podręczniku skrót **CM**) — model losowego systemu, który „nie ma pamięci": jego przyszłość zależy tylko od stanu obecnego, a nie od całej historii. To fundament dla: klasyfikacji stanów, rozkładów stacjonarnych, procesów kolejkowych i algorytmu Metropolis.

## Intuicja

Wyobraź sobie żabę skaczącą po $N$ kamieniach. Co sekundę żaba patrzy **tylko na kamień, na którym siedzi**, i losuje następny skok według ustalonych prawdopodobieństw. Nie pamięta, którędy przyszła. To jest własność Markowa — *mancanza di memoria* (brak pamięci).

**Proces stochastyczny** (wł. *processo stocastico*) to ogólnie rodzina zmiennych losowych $\{X(t)\}$ indeksowanych czasem $t$. W tym rozdziale czas jest **dyskretny** ($k = 0, 1, 2, \dots$) i przestrzeń stanów (wł. *spazio degli stati*) też jest dyskretna i skończona: $E = \{1, 2, \dots, N\}$.

Podręcznik podaje też przykłady procesów **z pamięcią** (których Markowem opisać się nie da): histereza magnetyczna, nieodwracalne odkształcenie materiału.

## 📘 Definicja 1.1 (łańcuch Markowa, s. 2 wydruku)

Rodzina zmiennych losowych $\{X_k\}$, $k \in \mathbb{N}$, o wartościach w $E=\{1,\dots,N\}$ jest **łańcuchem Markowa**, jeżeli

$$P(X_{n+1} = j \mid X_n = i,\ X_{n-1} = i_{n-1},\ \dots,\ X_0 = i_0) \;=\; P(X_{n+1} = j \mid X_n = i) \;=\; p_{ij}(n)$$

dla wszystkich stanów $i_0, \dots, i_{n-1}, i, j \in E$.

Symbol po symbolu:
- $X_n$ — stan systemu w chwili $n$ (zmienna losowa),
- lewa strona — prawdopodobieństwo przejścia do $j$ przy znajomości **całej historii**,
- prawa strona — to samo prawdopodobieństwo przy znajomości **tylko ostatniego stanu**,
- równość mówi: historia poza ostatnim stanem **nie wnosi nic** (własność Markowa),
- $p_{ij}(n)$ — **prawdopodobieństwo przejścia** (wł. *probabilità di transizione*) ze stanu $i$ do stanu $j$ w kroku $n$.

**Jednorodność w czasie** (wł. *temporalmente omogenea*): łańcuch jest jednorodny, gdy $p_{ij}(n) = p_{ij}$ nie zależy od $n$ — reguły gry są takie same w każdym kroku. W całym podręczniku zakładamy jednorodność.

Z własności Markowa wynika wzór na prawdopodobieństwo całej trajektorii (wzór (1.5) w podręczniku):

$$P(X_{n+1}=i_{n+1},\, X_n=i_n,\, \dots,\, X_0=i_0) = P(X_0=i_0)\; p_{i_0 i_1}\, p_{i_1 i_2}\cdots p_{i_n i_{n+1}}.$$

Czyli: prawdopodobieństwo ścieżki = (prawdopodobieństwo startu) razy (iloczyn prawdopodobieństw kolejnych skoków).

## 📘 Definicja 1.2 (macierz stochastyczna, s. 3 wydruku)

Macierz $A = (a_{ij})$, $i,j = 1,\dots,N$ jest **stochastyczna** (wł. *matrice stocastica*), jeżeli:
1. $a_{ij} \ge 0$ dla wszystkich $i, j$ (elementy nieujemne),
2. $\sum_{j=1}^{N} a_{ij} = 1$ dla każdego $i$ (**każdy wiersz sumuje się do 1**).

Macierz przejścia $P = (p_{ij})$ łańcucha jednorodnego jest stochastyczna, bo z $i$-tego stanu **na pewno gdzieś przejdziemy**:

$$\sum_{j=1}^N p_{ij} = P(X_{n+1} \in E \mid X_n = i) = 1.$$

I odwrotnie: każda macierz stochastyczna jest macierzą przejścia pewnego łańcucha Markowa.

### Jak czytać macierz $P$?

$$P = \begin{pmatrix} p_{11} & p_{12} & \cdots \\ p_{21} & p_{22} & \cdots \\ \vdots & & \end{pmatrix} \qquad \text{wiersz } i = \text{„jestem w stanie } i\text{: dokąd pójdę?''}$$

- **wiersz $i$** — rozkład prawdopodobieństwa następnego stanu, gdy teraz jestem w $i$,
- **kolumna $j$** — z jakich stanów (i z jakim prawd.) można wejść do $j$,
- $p_{ij} = P\{i \to j\}$ — przejście z $i$ (wiersz) do $j$ (kolumna) **w jednym kroku**.

## 📘 Esercizio 1.1 (s. 3 wydruku): iloczyn macierzy stochastycznych jest stochastyczny

**Teza:** jeśli $P$ i $Q$ są stochastyczne $N\times N$, to $PQ$ też.

**Rozwiązanie z podręcznika, krok po kroku.** Element iloczynu: $(PQ)_{ij} = \sum_{h=1}^N p_{ih} q_{hj} \ge 0$ (suma iloczynów liczb nieujemnych). Suma po wierszu $i$:

$$\sum_{j=1}^N (PQ)_{ij} = \sum_{j=1}^N \sum_{h=1}^N p_{ih} q_{hj} = \sum_{h=1}^N p_{ih} \underbrace{\sum_{j=1}^N q_{hj}}_{=1 \text{ ($Q$ stoch.)}} = \sum_{h=1}^N p_{ih} = 1.$$

(Zamieniliśmy kolejność sumowania — wolno, bo suma jest skończona.)

**Wniosek:** każda potęga $P^n = P \cdot P \cdots P$ jest macierzą stochastyczną. To będzie kluczowe w notatniku 03.

## 📘 Przykłady z podręcznika (s. 4–8 wydruku)

### Przykład 1.1 — schemat Bernoulliego
$E = \{1,2\}$, $p_{12} = p_{22} = p$, $p_{11} = p_{21} = q = 1-p$:

$$P = \begin{pmatrix} q & p \\ q & p \end{pmatrix}$$

Oba wiersze są identyczne — następny stan **nie zależy od obecnego**. Niezależne powtórzenia próby to szczególny (zdegenerowany) przypadek łańcucha Markowa.

### Przykład 1.2 — urządzenie sprawne/zepsute
$E = \{0, 1\}$ (0 = zepsute / wł. *guasto*, 1 = sprawne / wł. *funzionante*):

$$P = \begin{pmatrix} p & 1-p \\ 1-q & q \end{pmatrix}$$

Wiersz 0: zepsute zostaje zepsute z prawd. $p$, zostaje naprawione z prawd. $1-p$. Wiersz 1: sprawne psuje się z prawd. $1-q$.

### Przykład 1.3 — poziomy energetyczne elektronu
W modelu Bohra $p_{ij} = K\exp(-K\Delta_{ij})$, gdzie $\Delta_{ij} = E_j - E_i$ to zmiana energii. Morał: prawdopodobieństwo przejścia zależy tylko od pary (stan obecny, stan następny).

### Przykład 1.4 — automat stochastyczny
Automat skończony czyta litery z alfabetu $A=\{0,1\}$ i zmienia stan wg funkcji $f: E\times A \to E$. Gdy czytane litery $Z_n$ są **losowe i niezależne** ($P(Z_n{=}0)=P(Z_n{=}1)=\tfrac12$), ciąg stanów $X_{n+1} = f(X_n, Z_n)$ jest łańcuchem Markowa na $E=\{0,1,2,3\}$ z macierzą

$$P = \begin{pmatrix} 1/2 & 1/2 & 0 & 0 \\ 1/2 & 0 & 1/2 & 0 \\ 1/2 & 0 & 0 & 1/2 \\ 1/2 & 1/2 & 0 & 0 \end{pmatrix}$$

Automat z podręcznika rozpoznaje bloki trzech jedynek w strumieniu bitów — stan 3 oznacza „właśnie przeczytałem 111".

### Przykłady 1.5–1.7 — błądzenie losowe z barierami (random walk)

Cząstka na kracie $E = \{0, 1, \dots, a\}$ skacze o 1 w prawo z prawd. $p$, w lewo z prawd. $q = 1-p$ (nigdy nie stoi). Różne warianty różnią się zachowaniem na brzegach $0$ i $a$ (dla $a=3$):

**1.5 — bariery pochłaniające** (wł. *barriere assorbenti*): $p_{00} = 1$, $p_{aa} = 1$ — kto wpadnie, zostaje na zawsze („czarne dziury"):

$$P = \begin{pmatrix} 1 & 0 & 0 & 0 \\ q & 0 & p & 0 \\ 0 & q & 0 & p \\ 0 & 0 & 0 & 1 \end{pmatrix}$$

**1.6 — bariery odbijające** (wł. *barriere riflettenti*): $p_{01} = 1$, $p_{a,a-1} = 1$ — brzeg natychmiast odpycha:

$$P = \begin{pmatrix} 0 & 1 & 0 & 0 \\ q & 0 & p & 0 \\ 0 & q & 0 & p \\ 0 & 0 & 1 & 0 \end{pmatrix}$$

**1.7 — bariery częściowo odbijające**: na brzegu zostajemy z prawd. $r$, odbijamy się z prawd. $1-r$:

$$P = \begin{pmatrix} r & 1-r & 0 & 0 \\ q & 0 & p & 0 \\ 0 & q & 0 & p \\ 0 & 0 & 1-r & r \end{pmatrix}$$

Zauważ: wszystkie trzy macierze różnią się **tylko wierszami 0 i $a$** — bo tylko zachowanie brzegu się zmienia.

```python
import numpy as np

# 📘 Macierze z przykładów 1.5-1.7 dla a=3, p=0.5 — sprawdzamy stochastyczność
p, q, r = 0.5, 0.5, 0.3

P_abs = np.array([[1, 0, 0, 0],
                  [q, 0, p, 0],
                  [0, q, 0, p],
                  [0, 0, 0, 1]], dtype=float)   # bariery pochłaniające

P_odb = np.array([[0, 1, 0, 0],
                  [q, 0, p, 0],
                  [0, q, 0, p],
                  [0, 0, 1, 0]], dtype=float)   # bariery odbijające

P_cz = np.array([[r, 1-r, 0, 0],
                 [q, 0,   p, 0],
                 [0, q,   0, p],
                 [0, 0, 1-r, r]], dtype=float)  # częściowo odbijające

for nazwa, M in [("pochłaniające", P_abs), ("odbijające", P_odb), ("częściowe", P_cz)]:
    print(f"{nazwa:>15}: sumy wierszy = {M.sum(axis=1)}  (wszystkie 1 -> stochastyczna)")
```

## 🧪 Przykład własny: symulacja łańcucha i prawdopodobieństwo trajektorii

Weźmy urządzenie z Przykładu 1.2 z $p = 0.8$ (zepsute raczej zostaje zepsute), $q = 0.9$ (sprawne raczej działa). Policzmy ręcznie prawdopodobieństwo trajektorii $1 \to 1 \to 0 \to 0$ przy starcie $P(X_0 = 1) = 1$:

$$P(X_0{=}1, X_1{=}1, X_2{=}0, X_3{=}0) = 1 \cdot p_{11} \cdot p_{10} \cdot p_{00} = 0.9 \cdot 0.1 \cdot 0.8 = 0.072.$$

Poniżej potwierdzamy to symulacją Monte Carlo (ze stałym ziarnem — wyniki są powtarzalne).

```python
import numpy as np

rng = np.random.default_rng(seed=2026)   # stałe ziarno -> powtarzalny wynik

P = np.array([[0.8, 0.2],    # stan 0 (zepsute):  0->0: 0.8, 0->1: 0.2
              [0.1, 0.9]])   # stan 1 (sprawne):  1->0: 0.1, 1->1: 0.9

def symuluj(start, n_krokow):
    traj = [start]
    for _ in range(n_krokow):
        traj.append(rng.choice(2, p=P[traj[-1]]))  # losuj wg wiersza P
    return tuple(traj)

# prawdopodobieństwo trajektorii 1 -> 1 -> 0 -> 0
dokladnie = 1.0 * P[1,1] * P[1,0] * P[0,0]
N = 200_000
licznik = sum(symuluj(1, 3) == (1, 1, 0, 0) for _ in range(N))
print(f"teoria (iloczyn p_ij):    {dokladnie:.4f}")
print(f"symulacja ({N} prób):  {licznik/N:.4f}")
```

## Typowe błędy

1. **„Brak pamięci" nie znaczy „brak zależności"** — $X_{n+1}$ zależy od $X_n$! Nie zależy tylko od dalszej przeszłości.
2. **Wiersze vs kolumny**: u Abundo $p_{ij}$ to $i \to j$ i **wiersze** sumują się do 1. Zapis $\mu_{n+1} = \mu_n P$ (wektor wierszowy z lewej).
3. Mylenie jednorodności (reguły stałe w czasie) z własnością Markowa (brak pamięci) — to **dwa niezależne** założenia.
4. W przykładzie 1.1 oba wiersze są równe — to przypadek niezależności; w ogólnym łańcuchu wiersze są różne.

## Podsumowanie

- Łańcuch Markowa: przyszłość zależy tylko od teraźniejszości (Def. 1.1).
- Jednorodny: te same $p_{ij}$ w każdym kroku.
- Macierz przejścia $P$ jest wierszowo-stochastyczna (Def. 1.2); iloczyn macierzy stochastycznych jest stochastyczny (Es. 1.1).
- Prawdopodobieństwo trajektorii = start razy iloczyn $p_{ij}$ po drodze.
