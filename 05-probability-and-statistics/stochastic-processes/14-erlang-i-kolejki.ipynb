## Czego się uczymy i po co?

Zwieńczenie kursu: **teoria kolejek** — serce wydajności systemów komputerowych i telekomunikacji. Każda kolejka tutaj to proces narodzin i śmierci: „narodziny" = przyjście klienta (Poisson z intensywnością $\lambda$), „śmierć" = zakończenie obsługi (czasy wykładnicze z parametrem $\mu$).

## 📘 Problem Erlanga (§2.5)

$n$ okienek obsługi; klienci przychodzą wg procesu Poissona($\lambda$); czas obsługi $\sim Esp(\mu)$; klient zastający **wszystkie okienka zajęte jest odrzucany** (bez kolejki). Stan $k$ = liczba zajętych okienek.

Równania różniczkowe (2.43)–(2.45) wyprowadza się jak dla procesu narodzin i śmierci. W stanie stacjonarnym ($p_k' = 0$, $p_k \equiv \pi_k$) rozwiązanie to **wzory Erlanga** (2.48), z $\rho = \lambda/\mu$:

$$\pi_k = \frac{\rho^k/k!}{\sum_{j=0}^{n}\rho^j/j!}, \qquad k = 0, 1, \dots, n.$$

**Prawdopodobieństwo zablokowania** (wł. *probabilità di congestione*) — że wszystkie okienka zajęte i klient zostaje odrzucony — to $\pi_n$ (wzór (2.49), tzw. Erlang B). To proces narodzin i śmierci z $\lambda_k = \lambda$ dla $k<n$ (0 dalej) i $\mu_k = k\mu$ dla $k \le n$.

## 📘 Notacja Kendalla (§2.6)

Kolejkę opisuje zapis `pole1/pole2/pole3[/pole4/pole5]`:
- pole 1 — rozkład czasów między przyjściami (M = wykładniczy/markowowski, G = ogólny, D = deterministyczny),
- pole 2 — rozkład czasu obsługi,
- pole 3 — liczba stanowisk obsługi,
- pola 4, 5 — rozmiar bufora i liczba dopuszczanych klientów (pomijane, gdy nieskończone).

Stan systemu = **liczba klientów w systemie** (obsługiwani + czekający).

## 📘 Trzy kolejki (§2.6.1–2.6.3)

Wszystkie to procesy narodzin i śmierci z $\lambda_k = \lambda$; różni je tempo obsługi $\mu_k$. Z bilansu lokalnego ($\lambda\pi_k = \mu_{k+1}\pi_{k+1}$), $\rho = \lambda/\mu$:

**M/M/n** ($\mu_k = \min(k, n)\mu$): wzór (2.50)

$$\pi_k = \begin{cases} \frac{\rho^k}{k!}\pi_0, & k \le n \\ \frac{\rho^k}{n!\,n^{k-n}}\pi_0, & k \ge n+1 \end{cases}$$

Rozkład stacjonarny istnieje $\iff \rho < n$ (szereg geometryczny o ilorazie $\rho/n$); wtedy $\pi_0 = \left[\sum_{k=0}^n \frac{\rho^k}{k!} + \frac{\rho^{n+1}}{n!(n-\rho)}\right]^{-1}$ (2.55). Warunek $\rho < n$ mówi, **ile okienek trzeba otworzyć**, żeby kolejka się stabilizowała.

**M/M/∞** ($\mu_k = k\mu$): $\pi_k = \frac{\rho^k}{k!}e^{-\rho}$ — rozkład **Poissona($\rho$)**; równowaga istnieje zawsze (2.57).

**M/M/1** ($\mu_k = \mu$): $\pi_k = \rho^k(1-\rho)$ — rozkład **geometryczny**; wymaga $\rho < 1$ (2.60).

## 📘 Czas w systemie i w kolejce dla M/M/1 (§2.6.4–2.6.5)

Czas $T$ spędzony w systemie przez klienta (czekanie + własna obsługa): warunkowo na $N = n$ klientów zastanych $T \sim \Gamma(n+1, \mu)$; sumując po rozkładzie geometrycznym $N$ podręcznik oblicza:

$$P(T \le t) = 1 - e^{-\mu(1-\rho)t} \;\Longrightarrow\; T \sim Esp(\mu - \lambda), \qquad W := E(T) = \frac{1}{\mu - \lambda}.$$

Czas czekania $T_c$ w samej kolejce: $P(T_c = 0) = 1-\rho$ (system pusty — atom w zerze!), część ciągła o gęstości $\mu\rho(1-\rho)e^{-\mu(1-\rho)t}$; średnio

$$W_c := E(T_c) = \frac{\lambda}{\mu(\mu - \lambda)}, \qquad W = W_c + \frac{1}{\mu}.$$

## 📘 Relacje Little'a (§2.6.6)

Oznaczenia: $L$ — średnia liczba klientów w systemie, $L_c$ — w kolejce, $W$, $W_c$ — średnie czasy. **Prawa Little'a:**

$$L = \lambda W, \qquad L_c = \lambda W_c.$$

Dowód pierwszego w podręczniku: ergodyczność + lemat $X_t/t \to \lambda$ p.n. dla procesu Poissona (z prawa wielkich liczb). Dla M/M/1: $L = \frac{\lambda}{\mu-\lambda}$, $L_c = \frac{\lambda^2}{\mu(\mu-\lambda)}$ (2.75)–(2.76) — zgodnie z $W$, $W_c$ wyżej.

**Współczynnik wykorzystania** $\eta = \rho/n$: zawsze $\eta < 1$ w stochastycznym systemie stacjonarnym; gdy $\eta \to 1$, kolejki i czasy czekania **eksplodują do nieskończoności** — fundamentalny kompromis wydajności.

```python
import numpy as np
from math import factorial

# 📘 wzory Erlanga: n okienek, odrzucanie przy pełnym obłożeniu
def erlang_pi(rho, n):
    licz = [rho**k / factorial(k) for k in range(n + 1)]
    s = sum(licz)
    return [x / s for x in licz]

lam, mu, n = 4.0, 1.0, 6     # rho = 4
pi = erlang_pi(lam/mu, n)
print("pi_k:", np.round(pi, 4))
print(f"prawdopodobieństwo odrzucenia (Erlang B) pi_{n} = {pi[n]:.4f}")

# ile okienek, żeby odrzucać < 1% klientów?
for nn in range(1, 15):
    if erlang_pi(lam/mu, nn)[nn] < 0.01:
        print(f"wystarczy n = {nn} okienek (blokada {erlang_pi(lam/mu, nn)[nn]:.4f})"); break
```

```python
# 📘 M/M/1: rozkład geometryczny, W i prawa Little'a — weryfikacja symulacją
rng = np.random.default_rng(seed=1234)
lam, mu = 3.0, 4.0             # rho = 0.75 < 1
rho = lam / mu

# symulacja zdarzeniowa do czasu T_max
T_max, t, stan = 20_000.0, 0.0, 0
czas_w, pole_pod_N, przyjscia, czasy_klientow = {}, 0.0, 0, []
kolejka_wejsc = []
while t < T_max:
    tempo = lam + (mu if stan > 0 else 0)
    dt = rng.exponential(1 / tempo)
    czas_w[stan] = czas_w.get(stan, 0) + dt
    pole_pod_N += stan * dt
    t += dt
    if rng.random() < lam / tempo:       # przyjście klienta
        stan += 1; przyjscia += 1; kolejka_wejsc.append(t)
    else:                                 # zakończenie obsługi (FIFO)
        stan -= 1; czasy_klientow.append(t - kolejka_wejsc.pop(0))

total = sum(czas_w.values())
print("k | ułamek czasu | pi_k = rho^k (1-rho)")
for k in range(5):
    print(f"{k} |    {czas_w.get(k,0)/total:.4f}  | {rho**k*(1-rho):.4f}")
```

```python
# prawa Little'a i wzory jawne
L_sym = pole_pod_N / t
W_sym = np.mean(czasy_klientow)
lam_sym = przyjscia / t
print(f"L (symulacja) = {L_sym:.3f}   teoria lambda/(mu-lambda) = {lam/(mu-lam):.3f}")
print(f"W (symulacja) = {W_sym:.3f}   teoria 1/(mu-lambda)      = {1/(mu-lam):.3f}")
print(f"Little: L = lambda W ?  {L_sym:.3f} vs {lam_sym * W_sym:.3f}")

# czas w systemie ~ Esp(mu - lambda): sprawdzenie dystrybuanty w punkcie
tt = 1.0
emp = np.mean([x <= tt for x in czasy_klientow])
print(f"\nP(T <= 1): symulacja {emp:.4f}, teoria {1 - np.exp(-(mu-lam)*tt):.4f}")
```

```python
# eksplozja kolejki przy eta -> 1: L = lambda/(mu - lambda)
mu = 4.0
print("lambda |  rho  |   L (śr. liczba klientów)")
for lam_x in [2.0, 3.0, 3.6, 3.9, 3.99]:
    print(f" {lam_x:4} | {lam_x/mu:.3f} | {lam_x/(mu-lam_x):10.1f}")
# im bliżej pełnego wykorzystania, tym dramatyczniej rośnie kolejka
```

## Typowe błędy

1. **Erlang ≠ M/M/n**: w problemie Erlanga klient przy pełnym obłożeniu jest **odrzucany** (stanów jest $n+1$), w M/M/n — **czeka** (stany nieograniczone). Inne wzory!
2. Warunek stabilności: $\rho < 1$ dla M/M/1, $\rho < n$ dla M/M/n, **brak warunku** dla M/M/∞. Bez niego „rozkład stacjonarny" z wzorów nie istnieje.
3. $W$ (czas w systemie) vs $W_c$ (czas w kolejce): $W = W_c + 1/\mu$ — nie zapominaj o własnej obsłudze.
4. $T_c$ ma **atom w zerze** ($P(T_c = 0) = 1 - \rho$) — jego gęstość nie całkuje się do 1, tylko do $\rho$; to rozkład mieszany.
5. Prawa Little'a wiążą **średnie w reżimie stacjonarnym** — wymagają ergodyczności, nie działają w stanach przejściowych.

## Podsumowanie

- Kolejki M/M/… = procesy narodzin i śmierci; rozkłady stacjonarne z bilansu lokalnego.
- Erlang B: $\pi_n$ = prawd. odrzucenia przy $n$ okienkach bez kolejki.
- M/M/1: $\pi_k$ geometryczny; $T \sim Esp(\mu-\lambda)$; M/M/∞: $\pi_k$ Poissona; M/M/n: stabilna dla $\rho < n$.
- Little: $L = \lambda W$, $L_c = \lambda W_c$; $\eta \to 1$ $\Rightarrow$ kolejki eksplodują.

## Pytania kontrolne

1. Serwer obsługuje średnio 10 żądań/s ($\mu = 10$), żądania przychodzą z $\lambda = 8$/s. Ile wynosi średni czas odpowiedzi $W$ i średnia liczba żądań w systemie $L$?
2. Dlaczego kolejka M/M/∞ jest stabilna dla każdego $\rho$?
3. W M/M/1 z $\rho = 0.9$: jakie jest prawdopodobieństwo, że przychodzący klient w ogóle nie będzie czekał?

*(Odpowiedzi poniżej.)*

## Odpowiedzi

1. $W = 1/(\mu - \lambda) = 1/2$ s; $L = \lambda W = 8 \cdot 0.5 = 4$ żądania (albo wprost $\lambda/(\mu-\lambda) = 4$).
2. Bo tempo obsługi $\mu_k = k\mu$ rośnie liniowo ze stanem — im więcej klientów, tym szybciej system się opróżnia; szereg $\sum \rho^k/k!$ zbiega zawsze (to szereg $e^\rho$).
3. $P(T_c = 0) = \pi_0 = 1 - \rho = 0.1$ — tylko 10% klientów trafia na pusty system.
