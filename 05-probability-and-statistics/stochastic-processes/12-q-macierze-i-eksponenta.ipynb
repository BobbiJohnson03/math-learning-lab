## Czego się uczymy i po co?

W czasie dyskretnym łańcuch opisywała **macierz przejścia** $P$. W czasie ciągłym kroku „co 1" nie ma — zamiast tego system opisuje macierz **temp przejść** $Q$ (generator), a prawdopodobieństwa przejścia w czasie $t$ to $P(t) = e^{tQ}$. Nauczymy się: czym jest $Q$-macierz, jak działa eksponenta macierzy i jak obie macierze są powiązane.

### Najważniejsze rozróżnienie: $Q$ vs $P(t)$

| | $Q$ (generator) | $P(t)$ (macierz przejścia) |
|---|---|---|
| zawartość | **tempa** (1/czas) | **prawdopodobieństwa** |
| przekątna | $q_{ii} \le 0$ | $p_{ii}(t) \in [0,1]$ |
| poza przekątną | $q_{ij} \ge 0$ | $p_{ij}(t) \ge 0$ |
| suma wiersza | $0$ | $1$ |
| związek | $Q = P'(0)$ | $P(t) = e^{tQ}$ |

## 📘 Definicja 2.2 ($Q$-macierz, s. 46 wydruku)

Macierz rzeczywista $Q = (q_{ij})$ na przestrzeni stanów $I$ jest **$Q$-macierzą**, jeżeli:

$$\text{(i)}\ q_{ii} \le 0, \qquad \text{(ii)}\ q_{ij} \ge 0 \ (i \ne j), \qquad \text{(iii)}\ \sum_j q_{ij} = 0 \ \forall i.$$

Z (iii): $-q_{ii} = \sum_{j \ne i} q_{ij}$. Interpretacja:
- $q_{ij}$ ($i\ne j$) — **tempo przejścia** z $i$ do $j$,
- $q_i := -q_{ii}$ — **tempo opuszczania** stanu $i$ (suma temp wyjść we wszystkie strony).

Przykłady z podręcznika: macierz zerowa (2.1); ogólna $Q$-macierz $2\times2$ (2.2): $Q = \begin{pmatrix} -\alpha & \alpha \\ \beta & -\beta \end{pmatrix}$, $\alpha,\beta \ge 0$.

## 📘 Definicja 2.3 (eksponenta macierzy)

$$e^{tQ} = I + \sum_{k=1}^{\infty} \frac{(tQ)^k}{k!}$$

— dosłownie szereg $e^x$ z macierzą w roli $x$. Szereg zbiega zawsze (oszacowanie normą: $\|e^{tQ}\| \le e^{|t|\|Q\|}$).

### 📘 Własności (s. 47)

$$\text{(I)}\quad e^{tQ}e^{sQ} = e^{(t+s)Q} \qquad \text{(II)}\quad \frac{d}{dt}e^{tQ} = Qe^{tQ} = e^{tQ}Q \qquad \text{(III)}\quad \det(e^{tQ}) = e^{t\,\mathrm{Tr}Q}$$

($\mathrm{Tr}Q = \sum_i q_{ii}$ — ślad). Z (I): $e^{-tQ} = (e^{tQ})^{-1}$ — eksponenta jest zawsze odwracalna. ⚠️ Dla **różnych** macierzy $A, B$ na ogół $e^{A+B} \ne e^A e^B$ (chyba że komutują) — własność (I) działa, bo $tQ$ i $sQ$ komutują.

## 📘 Teorema 2.1: równania Kołmogorowa dla $P(t) = e^{tQ}$

Dla skończonej $Q$-macierzy rodzina $P(t) = e^{tQ}$, $t \ge 0$ spełnia:

**(a) własność półgrupy:** $P(t+s) = P(s)P(t)$ — przejście w czasie $t+s$ to złożenie przejść (ciągły odpowiednik $P^{n+m} = P^nP^m$);

**(b)** $P(t)$ jest **jedynym** rozwiązaniem równań różniczkowych z $P(0) = I$:

$$\frac{d}{dt}P(t) = P(t)\,Q \quad \text{(forward)}, \qquad \frac{d}{dt}P(t) = Q\,P(t) \quad \text{(backward)};$$

**(c)** $\frac{d^n}{dt^n}P(t)\big|_{t=0} = Q^n$; w szczególności $P'(0) = Q$ — **generator to pochodna macierzy przejścia w zerze**.

Dowód (a) to wzór dwumianowy Newtona + przestawienie sum; jedyność w (b) przez pokazanie, że $Z(t)e^{-tQ}$ ma pochodną zero.

## 📘 Teorema 2.2: charakteryzacja

$P(t) = e^{tQ}$ jest macierzą **stochastyczną dla każdego $t \ge 0$** $\iff$ $Q$ jest $Q$-macierzą.

Intuicja „$\Leftarrow$": dla małych $h$ mamy $P(h) = I + hQ + o(h)$, czyli $p_{ii}(h) = 1 + q_{ii}h + o(h) > 0$ i $p_{ij}(h) = q_{ij}h + o(h) \ge 0$, a sumy wierszy $= 1$ bo sumy wierszy $Q^k$ są 0; dla dużych $t$ używamy własności półgrupy: $P(t) = (P(t/n))^n$.

### 📘 Obserwacje (2.6–2.9, s. 50–53)

- *Oss. 2.6:* jeśli dla każdej pary $i \ne j$ istnieje $n$ z $(Q^n)_{ij} > 0$, to $P(h) > 0$ — macierz regularna.
- *Oss. 2.7:* $P(t)$ bistochastyczna $\iff$ kolumny $Q$ sumują się do zera.
- *Oss. 2.8:* jeśli $\gamma$ jest wartością własną $Q$, to $e^{t\gamma}$ jest wartością własną $P(t)$. $Q$ ma zawsze wartość własną $0$ (odpowiednik $\lambda = 1$ dla $P$!). Ponadto $\det P(t) = e^{t\mathrm{Tr}Q} > 0$ — $P(t)$ nigdy nie jest osobliwa.
- *Oss. 2.9:* (1) $Q$-macierz jest zawsze osobliwa ($\det Q = 0$); (2) $Q^2$ na ogół **nie jest** $Q$-macierzą; (4) nie każda macierz stochastyczna $P$ jest postaci $e^{tQ}$ — np. mająca $\det P \le 0$ (bo $\det e^{tQ} > 0$ zawsze).

## 📘 Przykład 2.3: jawne $P(t)$ dla dwóch stanów (s. 51 wydruku)

$Q = \begin{pmatrix} -\alpha & \alpha \\ \beta & -\beta \end{pmatrix}$, wartości własne $0$ i $-(\alpha+\beta)$. Diagonalizacja $Q = U\Lambda U^{-1}$ daje $P(t) = Ue^{t\Lambda}U^{-1}$, a elementy mają postać $p_{ij}(t) = A_{ij} + B_{ij}e^{-(\alpha+\beta)t}$ ze stałymi wyznaczonymi z $P(0) = I$ i $P'(0) = Q$:

$$P(t) = \begin{pmatrix} \frac{\beta}{\alpha+\beta} + \frac{\alpha}{\alpha+\beta}e^{-(\alpha+\beta)t} & \frac{\alpha}{\alpha+\beta} - \frac{\alpha}{\alpha+\beta}e^{-(\alpha+\beta)t} \\ \frac{\beta}{\alpha+\beta} - \frac{\beta}{\alpha+\beta}e^{-(\alpha+\beta)t} & \frac{\alpha}{\alpha+\beta} + \frac{\beta}{\alpha+\beta}e^{-(\alpha+\beta)t} \end{pmatrix} \xrightarrow{t\to\infty} \begin{pmatrix} \frac{\beta}{\alpha+\beta} & \frac{\alpha}{\alpha+\beta} \\ \frac{\beta}{\alpha+\beta} & \frac{\alpha}{\alpha+\beta} \end{pmatrix}$$

Granica ma równe wiersze — to rozkład stacjonarny $\pi = \left(\frac{\beta}{\alpha+\beta}, \frac{\alpha}{\alpha+\beta}\right)$ (pełna analogia ze wzorem (1.44) dla czasu dyskretnego!).

## 📘 Przykład 2.4: wartości własne wielokrotne

$Q$ (4×4, cykliczna z tempami) ma widmo $0, -2, -3, -3$ — przy podwójnej wartości własnej o krotności geometrycznej 1 elementy $P(t)$ zawierają człony $t\,e^{-3t}$:

$$p_{ij}(t) = A_{ij} + B_{ij}e^{-2t} + (C_{ij} + D_{ij}t)\,e^{-3t}.$$

To samo zjawisko co „$n\lambda^n$" przy niediagonalizowalnych macierzach w czasie dyskretnym.

```python
import numpy as np
from scipy.linalg import expm   # eksponenta macierzy (odpowiednik expm() w Scilabie)

# 📘 Przykład 2.3: porównanie expm z wzorem jawnym
al, be = 0.7, 0.3
Q = np.array([[-al, al],
              [be, -be]])
print("Q-macierz? sumy wierszy:", Q.sum(axis=1))

def P_jawne(t):
    s = al + be
    return np.array([[be/s + al/s*np.exp(-s*t), al/s - al/s*np.exp(-s*t)],
                     [be/s - be/s*np.exp(-s*t), al/s + be/s*np.exp(-s*t)]])

for t in [0.1, 1.0, 5.0]:
    r = abs(expm(t*Q) - P_jawne(t)).max()
    print(f"t={t}: max|expm - wzór| = {r:.2e}")
print("\nP(100) ~ granica (równe wiersze pi):")
print(np.round(expm(100*Q), 6), "\npi teoretyczne:", (be/(al+be), al/(al+be)))
```

```python
# własności (I)-(III)
t, s = 0.6, 1.1
print("(I)  P(t)P(s) == P(t+s):", np.allclose(expm(t*Q) @ expm(s*Q), expm((t+s)*Q)))

h = 1e-7   # (II) pochodna numeryczna w t: dP/dt = Q P(t)
dP = (expm((t+h)*Q) - expm(t*Q)) / h
print("(II) dP/dt == Q P(t):   ", np.allclose(dP, Q @ expm(t*Q), atol=1e-5))

print("(III) det P(t) =", np.linalg.det(expm(t*Q)),
      " e^(t TrQ) =", np.exp(t*np.trace(Q)))

# Oss. 2.8: wartości własne P(t) to e^(t*gamma)
gammy = np.linalg.eigvals(Q)
print("\nwidmo Q:", np.round(gammy, 4), " -> widmo P(1):", np.round(np.exp(gammy), 4))
print("widmo P(1) z eig:", np.round(np.linalg.eigvals(expm(Q)), 4))
```

```python
# 📘 Przykład 2.4: widmo z wartością podwójną
Q4 = np.array([[-2, 1, 1, 0],
               [1, -2, 1, 0],
               [0, 1, -2, 1],
               [1, 0, 1, -2]], dtype=float)
print("sumy wierszy:", Q4.sum(axis=1))
print("widmo Q4:", np.round(sorted(np.linalg.eigvals(Q4).real), 4), " (0, -2, -3, -3)")
print("\nP(1) = expm(Q4):")
print(np.round(expm(Q4), 5))
print("sumy wierszy P(1):", expm(Q4).sum(axis=1))
```

## Typowe błędy

1. **Mylenie $Q$ z $P(t)$**: wiersze $Q$ sumują się do **0** (tempa), wiersze $P(t)$ do **1** (prawdopodobieństwa); przekątna $Q$ jest $\le 0$, przekątna $P(t)$ — nieujemna.
2. `expm(Q)` to **nie** `np.exp(Q)`! Pierwsza to eksponenta macierzowa (szereg potęg macierzy), druga — eksponenta po elementach. W Scilabie analogicznie: `expm()` vs `exp()`.
3. $e^{A}e^{B} \ne e^{A+B}$ dla niekomutujących macierzy; wolno tylko dla $tQ, sQ$ (ta sama macierz).
4. Nie każda macierz stochastyczna ma postać $e^{tQ}$ (np. z $\det \le 0$) — czas ciągły „nie umie" wygenerować wszystkich przejść dyskretnych.
5. $Q$ ma zawsze $\det Q = 0$ — nie próbuj jej odwracać.

## Podsumowanie

- $Q$-macierz: przekątna $\le 0$, poza nią $\ge 0$, wiersze sumują się do 0; $q_i = -q_{ii}$ to tempo opuszczania stanu.
- $P(t) = e^{tQ}$ (szereg); własność półgrupy; równania forward $P' = PQ$ i backward $P' = QP$; $P'(0) = Q$.
- $P(t)$ stochastyczna dla wszystkich $t$ $\iff$ $Q$ jest $Q$-macierzą.
- Widmo: $\gamma \mapsto e^{t\gamma}$; $0$ u $Q$ odpowiada $1$ u $P(t)$.

## Odpowiedzi
