## Czego się uczymy i po co?

Nie wszystkie stany są sobie równe: z niektórych łańcuch wraca w nieskończoność, inne odwiedza tylko „przejazdem", a z niektórych nie ma już wyjścia. Klasyfikacja stanów to **mapa terenu** — bez niej nie zrozumiemy problemów pochłaniania (notatnik 05) ani rozkładów stacjonarnych (notatnik 06).

## 📘 Definicje (§1.4, s. 14 wydruku)

Niech $\{X_k\}$ będzie jednorodnym łańcuchem na $E = \{1,\dots,N\}$ z macierzą $P$.

1. Stan $i$ **komunikuje się** z $j$ (zapis $i \rightharpoonup j$), jeśli istnieje $n > 0$ takie, że $p_{ij}^{(n)} > 0$ — da się dojść z $i$ do $j$ w pewnej liczbie kroków z dodatnim prawdopodobieństwem. *(Uwaga: w tym podręczniku relacja jest jednostronna — $i \rightharpoonup j$ nie wymaga $j \rightharpoonup i$.)*
2. Podzbiór $C \subset E$ to **klasa zamknięta** (wł. *classe chiusa*), jeśli ze stanów $C$ nie można się skomunikować z resztą $E$ — „wejście bez wyjścia".
3. Klasa zamknięta jest **nieprzywiedlna** (wł. *irriducibile*), jeśli wszystkie jej stany komunikują się nawzajem.
4. Stan tworzący samodzielnie klasę nieprzywiedlną to stan **pochłaniający** (wł. *assorbente*): $p_{hh} = 1$.
5. Łańcuch jest **nieprzywiedlny**, jeśli całe $E$ to jedna klasa nieprzywiedlna (zewsząd dojdziesz wszędzie).
6. $\tau_j = \min\{n > 0: X_n = j\}$ — chwila **pierwszej wizyty** w $j$ (może być $+\infty$). Definiujemy $\rho_{ij} = P(\tau_j < \infty \mid X_0 = i)$ — prawd., że startując z $i$, kiedykolwiek odwiedzimy $j$.
7. Stan $i$ jest **chwilowy** (wł. *transitorio*), gdy $\rho_{ii} < 1$ (możliwy brak powrotu), a **powracający** (wł. *ricorrente*), gdy $\rho_{ii} = 1$ (powrót pewny, a więc nieskończenie wiele powrotów).

### Kryterium przez szereg (s. 15–16 wydruku)

Licząc średnią liczbę wizyt w $i$ (przez zmienne wskaźnikowe $I_n$), podręcznik pokazuje:

$$i \text{ powracający} \iff \sum_{n=0}^{\infty} p_{ii}^{(n)} = \infty, \qquad\qquad i \text{ chwilowy} \iff \sum_{n=0}^{\infty} p_{ii}^{(n)} < \infty.$$

### 📘 Osservazione 1.1 (przechodniość komunikacji)

Jeśli $i \rightharpoonup j$ i $j \rightharpoonup h$, to $i \rightharpoonup h$. Dowód: $p_{ih}^{(n+m)} \ge p_{ij}^{(n)}\, p_{jh}^{(m)} > 0$ — można iść przez $j$.

### 📘 Esercizi 1.3–1.4 (do samodzielnej pracy)

- 1.3: jeśli $i$ komunikuje się ze stanem chwilowym $j$ *(i wzajemnie — por. kontekst)*, to $i$ też jest chwilowy.
- 1.4: jeśli powracający $i$ komunikuje się z $j$ i $j$ z $i$, to $j$ też jest powracający. Morał: **powracalność/chwilowość to własność całej klasy**.

## 📘 Przykład 1.8-bis: duża macierz $10\times 10$ (s. 15 wydruku)

*(Uwaga: w podręczniku są dwa przykłady o numerze 1.8 — ruina gracza na s. 8 i ten; to drobna usterka numeracji źródła.)*

Macierz podana jest schematycznie: $*$ oznacza element dodatni. Analizując, które przejścia są możliwe, podręcznik znajduje **rozkład przestrzeni stanów**:

$$E = T \cup C_1 \cup C_2 \cup C_3, \qquad C_1 = \{1,7,9\},\quad C_2 = \{2,4\},\quad C_3 = \{10\},\quad T = \{3,5,6,8\}.$$

- $C_1, C_2$ — klasy nieprzywiedlne stanów powracających,
- $C_3 = \{10\}$ — stan pochłaniający,
- $T$ — stany chwilowe (np. $3 \rightharpoonup 5 \rightharpoonup 2$, ale z 2 nie ma powrotu do 3).

**Twierdzenie o rozkładzie:** każdy skończony łańcuch rozkłada się jednoznacznie na $E = T \cup C_1 \cup C_2 \cup \dots$, gdzie $C_i$ to nieprzywiedlne klasy zamknięte, a $T$ — zbiór stanów chwilowych.

### Ważna obserwacja z podręcznika (s. 16)

Mogłoby się wydawać, że stan $1 \in C_1$ jest chwilowy, bo istnieje trajektoria $1 \to 9 \to 9 \to 9 \to \cdots$, która nigdy nie wraca do 1. Ale prawdopodobieństwo **pozostania w 9 na zawsze** to $\lim_n (p_{99})^n = 0$ (bo $p_{99} < 1$). Pojedyncza „zła" trajektoria o prawdopodobieństwie zero nie czyni stanu chwilowym — liczy się, czy $\rho_{11} < 1$.

## 📘 Przykład 1.9 — ruina gracza

W łańcuchu ruiny gracza ($0 < p < 1$): stany $0$ i $a$ pochłaniające (powracające), stany $1, \dots, a-1$ chwilowe (bo $i \rightharpoonup 0$, ale z 0 nie wrócimy).

### Czas pierwszej wizyty (s. 16)

Jeżeli $j$ chwilowy, to $P(\tau_j = +\infty \mid X_0 = j) > 0$ i wtedy przyjmujemy $E(\tau_j) = +\infty$. Jeżeli $P(\tau_j = \infty) = 0$, to $E(\tau_j) = \sum_k k\,P(\tau_j = k)$ — skończona, gdy szereg zbieżny.

## 🧪 Przykład własny: klasyfikacja małego łańcucha w kodzie

Weźmy łańcuch o 5 stanach (własna konstrukcja — prostsza niż $10\times 10$ z podręcznika, ale z tą samą strukturą: klasa nieprzywiedlna, stan pochłaniający, stany chwilowe):

$$P = \begin{pmatrix} 0.5 & 0.5 & 0 & 0 & 0 \\ 0.3 & 0.7 & 0 & 0 & 0 \\ 0.2 & 0 & 0.3 & 0.5 & 0 \\ 0 & 0 & 0.4 & 0.1 & 0.5 \\ 0 & 0 & 0 & 0 & 1 \end{pmatrix}$$

Oczekiwania: $C_1 = \{0, 1\}$ klasa zamknięta (powracające), $C_2 = \{4\}$ pochłaniający, $T = \{2, 3\}$ chwilowe. Sprawdzimy algorytmicznie przez osiągalność (domknięcie przechodnie grafu przejść).

```python
import numpy as np

P = np.array([[0.5, 0.5, 0,   0,   0  ],
              [0.3, 0.7, 0,   0,   0  ],
              [0.2, 0,   0.3, 0.5, 0  ],
              [0,   0,   0.4, 0.1, 0.5],
              [0,   0,   0,   0,   1  ]])
N = P.shape[0]

# osiągalność: czy istnieje n>0 z p_ij^(n) > 0?  Liczymy domknięcie przechodnie.
dosiegalne = (P > 0)
for _ in range(N):   # wystarczy N iteracji
    dosiegalne = dosiegalne | (dosiegalne @ dosiegalne)

# stan i jest powracający <=> dla każdego j osiągalnego z i, także i jest osiągalne z j
powracajace, chwilowe = [], []
for i in range(N):
    if all((not dosiegalne[i, j]) or dosiegalne[j, i] for j in range(N)):
        powracajace.append(i)
    else:
        chwilowe.append(i)

print("stany powracające:", powracajace)    # oczekujemy [0, 1, 4]
print("stany chwilowe:   ", chwilowe)       # oczekujemy [2, 3]
print("stan pochłaniający:", [i for i in range(N) if P[i, i] == 1])
```

```python
# ilustracja kryterium szeregowego: suma p_ii^(n) rośnie dla powracających,
# a dla chwilowych zbiega do skończonej granicy
suma = np.zeros(N)
Pn = np.eye(N)
for n in range(200):
    suma += np.diag(Pn)
    Pn = Pn @ P
for i in range(N):
    rodzaj = "powracający" if i in powracajace else "chwilowy"
    print(f"stan {i} ({rodzaj:11s}): suma p_ii^(n) dla n<200 = {suma[i]:9.3f}")
# dla chwilowych wartość praktycznie przestaje rosnąć — szereg zbieżny
```

## Typowe błędy

1. „Komunikuje się" u Abundo jest **jednostronne** ($i \rightharpoonup j$); w wielu podręcznikach „communicate" oznacza relację obustronną. Czytaj definicje lokalnie!
2. Istnienie trajektorii bez powrotu **nie** oznacza chwilowości — liczy się prawdopodobieństwo (obserwacja przy przykładzie $10\times10$).
3. Stan pochłaniający to szczególny stan powracający (wraca w każdym kroku), nie osobna kategoria.
4. W skończonym łańcuchu **nie wszystkie** stany mogą być chwilowe — łańcuch musi gdzieś „osiąść"; zawsze istnieje przynajmniej jedna klasa powracająca.

## Podsumowanie

- $i \rightharpoonup j$: osiągalność z dodatnim prawdopodobieństwem w pewnej liczbie kroków.
- Rozkład $E = T \cup C_1 \cup C_2 \cup \cdots$: klasy nieprzywiedlne + stany chwilowe.
- Powracający: $\rho_{ii} = 1 \iff \sum_n p_{ii}^{(n)} = \infty$; chwilowy: $\rho_{ii} < 1 \iff$ suma skończona.
- Powracalność to własność klasy (Esercizi 1.3–1.4).
