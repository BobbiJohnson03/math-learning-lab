## Czego się uczymy i po co?

Wchodzimy w **czas ciągły**: zdarzenia (zgłoszenia do serwera, awarie, klienci) przychodzą w losowych chwilach. Proces Poissona to najprostszy i najważniejszy model takiego „strumienia zdarzeń" — będzie cegiełką wszystkich procesów kolejkowych (notatnik 14).

### Nowe obiekty

- $N(t)$ — liczba zdarzeń do chwili $t$ (proces zliczający, wł. *processo di conteggio*), $N(0) = 0$;
- $\tau_i$ — chwila $i$-tego zdarzenia;
- $T_i = \tau_i - \tau_{i-1}$ — **czasy międzyprzyjść** (wł. *intertempi di arrivo*);
- $\lambda > 0$ — **intensywność** (średnia liczba zdarzeń na jednostkę czasu).

## 📘 Konstrukcja I: z czasów międzyprzyjść (s. 38 wydruku)

**Założenie:** $T_1, T_2, \dots$ są niezależne, każdy o rozkładzie wykładniczym $Esp(\lambda)$ z gęstością $f(t) = \lambda e^{-\lambda t}$ dla $t \ge 0$.

Chwila $k$-tego zdarzenia to suma $\tau_k = T_1 + \cdots + T_k$. Z twierdzenia o sumie niezależnych zmiennych Gamma ($Esp(\lambda) = \Gamma(1,\lambda)$):

$$\tau_k \sim \Gamma(k, \lambda), \qquad g(t) = \frac{\lambda^k}{\Gamma(k)}\,t^{k-1}e^{-\lambda t},\ t\ge0,$$

gdzie $\Gamma(k) = (k-1)!$ dla $k$ całkowitego.

Kluczowa tożsamość zdarzeń: $\{N(T) \le k\} = \{\tau_{k+1} > T\}$ („do chwili $T$ co najwyżej $k$ zdarzeń" = „$(k{+}1)$-sze zdarzenie po $T$"). Z dystrybuanty rozkładu Gamma (wzór (2.4)):

$$P(N(T) \le k) = \sum_{i=0}^{k} \frac{(\lambda T)^i}{i!}e^{-\lambda T}$$

— a to jest dystrybuanta rozkładu **Poissona** o parametrze $\lambda T$. Stąd:

## 📘 Definicja 2.1

**Proces Poissona o intensywności $\lambda > 0$** to proces $N(t)$ o wartościach całkowitych $\ge 0$ z

$$P(N(t) = k) = e^{-\lambda t}\,\frac{(\lambda t)^k}{k!}, \qquad k \ge 0.$$

Trajektoria to niemalejąca funkcja schodkowa: w chwilach $\tau_k$ skok o 1. Z własności rozkładu Poissona: $E(N(t)) = Var(N(t)) = \lambda t$ (Osservazione 2.1).

## 📘 Konstrukcja II: aksjomatyczna (Proposizione 2.1, s. 39)

Proces zliczający $X(t)$ jest procesem Poissona o intensywności $\lambda$, jeżeli:
1. przyrosty na rozłącznych przedziałach są **niezależne**;
2. $X(s+t) - X(s)$ zależy tylko od $t$ (**jednorodność**);
3. $X(0) = 0$;
4. ze stanu $j$ można przejść bezpośrednio tylko do $j+1$;
5. $P(\text{skok w } [t, t+h]) = \lambda h + o(h)$;
6. $P(\text{2 lub więcej skoków w } [t, t+h]) = o(h)$.

Symbol $o(h)$ („o małe") oznacza wielkość znikającą szybciej niż $h$: $o(h)/h \to 0$ przy $h\to0$.

**Szkic dowodu z podręcznika** (warto prześledzić — ta technika wróci w notatnikach 11 i 14): rozbijamy $\{X(t+h) = n\}$ na trzy scenariusze (brak skoku / jeden skok / więcej), co daje

$$p_n(t+h) = p_n(t)(1 - \lambda h) + p_{n-1}(t)\lambda h + o(h),$$

dzielimy przez $h$, przechodzimy do granicy $h \to 0$ i dostajemy **układ równań różniczkowych** (2.8):

$$p_n'(t) = -\lambda p_n(t) + \lambda p_{n-1}(t),$$

którego rozwiązaniem (z $p_0(0)=1$) jest $p_n(t) = e^{-\lambda t}(\lambda t)^n/n!$ — rozkład Poissona. ✔

## 📘 Dwa wnioski (s. 40–42)

**Osservazione 2.2:** z konstrukcji II można odzyskać konstrukcję I — czasy międzyprzyjść są niezależne i wykładnicze. Np. $P(T_k > t) = P(X(\tau_k + t) - X(\tau_k) = 0) = P(X(t) = 0) = e^{-\lambda t}$.

**Zastosowanie:** jeżeli w $[0, T]$ zaszło **dokładnie jedno** zdarzenie, to chwila jego zajścia ma rozkład **jednostajny** na $[0, T]$:

$$P(\tau \le t \mid N(T) = 1) = \frac{\lambda t e^{-\lambda t}\cdot e^{-\lambda(T-t)}}{\lambda T e^{-\lambda T}} = \frac{t}{T}.$$

Poissonowskie zdarzenia „nie preferują" żadnej części przedziału.

```python
import numpy as np

# symulacja procesu Poissona z czasów międzyprzyjść (konstrukcja I)
rng = np.random.default_rng(seed=314)
lam, T = 2.0, 10.0          # intensywność 2 zdarzenia / jednostkę czasu

def trajektoria():
    t, chwile = 0.0, []
    while True:
        t += rng.exponential(1 / lam)     # T_i ~ Esp(lambda)
        if t > T: break
        chwile.append(t)
    return chwile

# sprawdzenie: N(T) ~ Poisson(lambda*T) = Poisson(20)
proby = [len(trajektoria()) for _ in range(20_000)]
print(f"E[N(T)]  symulacja: {np.mean(proby):.3f}   teoria: {lam*T}")
print(f"Var[N(T)] symulacja: {np.var(proby):.3f}   teoria: {lam*T}")
```

```python
from math import exp, factorial

# porównanie rozkładu empirycznego z teoretycznym Poissonem(20)
teoria = lambda k: exp(-lam*T) * (lam*T)**k / factorial(k)
print(" k | empirycznie | teoria")
for k in range(15, 26, 2):
    emp = np.mean([x == k for x in proby])
    print(f"{k:2d} |    {emp:.4f}   | {teoria(k):.4f}")
```

```python
# zastosowanie: jeśli N(T)=1, chwila zdarzenia ~ U(0, T)
rng = np.random.default_rng(seed=314)
chwile_pojedynczych = []
for _ in range(100_000):
    tr = trajektoria()
    if len(tr) == 1:
        chwile_pojedynczych.append(tr[0])
ch = np.array(chwile_pojedynczych)
print(f"liczba trajektorii z N(T)=1: {len(ch)}")
print(f"średnia chwila: {ch.mean():.3f}  (U(0,10) -> 5.0)")
print(f"P(chwila < 5): {(ch < 5).mean():.3f}  (teoria 0.5)")
```

## Typowe błędy

1. Mylenie $\lambda$ (zdarzenia na jednostkę czasu) z $1/\lambda$ (średni czas **między** zdarzeniami). $E(T_i) = 1/\lambda$, a $E(N(t)) = \lambda t$.
2. $o(h)$ to nie „jakaś mała liczba" — to funkcja znikająca szybciej niż $h$; pisząc $\lambda h + o(h)$ twierdzimy coś o granicy $h\to0$.
3. Rozkład Poissona (dyskretny — liczba zdarzeń) a rozkład wykładniczy (ciągły — czas między zdarzeniami) to **dwie strony tego samego procesu**; nie mylić.
4. Brak pamięci wykładniczego: $P(T > s + t \mid T > s) = P(T > t)$ — „stare" oczekiwanie nie skraca pozostałego czasu. Bez tej własności konstrukcja się załamuje.

## Podsumowanie

- Proces Poissona = zliczanie zdarzeń z niezależnymi, wykładniczymi czasami międzyprzyjść.
- $N(t) \sim Poisson(\lambda t)$; $E = Var = \lambda t$.
- Równoważna definicja aksjomatyczna: niezależne przyrosty + $P(\text{skok}) = \lambda h + o(h)$.
- Przy $N(T) = 1$ chwila zdarzenia jest jednostajna na $[0,T]$.

## Odpowiedzi

1. $P(N(2) = 0) = e^{-3\cdot2} = e^{-6} \approx 0.0025$.
2. $\tau_3 = T_1+T_2+T_3 \sim \Gamma(3, \lambda)$; $E(\tau_3) = 3/\lambda = 1$ min.
3. Bo $P(N(T)\le k) = P(\tau_{k+1} > T) = 1 - F_{\Gamma(k+1,\lambda)}(T)$, a ta wielkość (wzór (2.4)) równa się $\sum_{i=0}^k \frac{(\lambda T)^i}{i!}e^{-\lambda T}$ — dystrybuancie Poissona($\lambda T$). Skoro dystrybuanty się pokrywają, rozkłady są równe.
