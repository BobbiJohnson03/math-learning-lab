## Czego się uczymy i po co?

Składamy wszystko w całość: definiujemy łańcuch Markowa w czasie ciągłym (CTMC), uczymy się znajdować jego rozkład stacjonarny ($\pi Q = 0$ zamiast $\pi P = \pi$!) i rozumieć jego mechanikę: **wykładnicze czasy pobytu + dyskretny łańcuch skoków**.

## 📘 Definicja (s. 54 wydruku)

**CM z czasem ciągłym, dyskretną przestrzenią stanów $I$, jednorodna, o generatorze $Q$ i rozkładzie początkowym $\nu$** (krótko: $(\nu, Q)$-CM) to rodzina $\{X_t,\ t\ge0\}$ taka, że:

(a) $P(X_0 = i) = \nu_i$;

(b) dla $0 < t_1 < \dots < t_n$:

$$P(X_0 = i_0, X_{t_1} = i_1, \dots, X_{t_n} = i_n) = \nu_{i_0}\, p_{i_0i_1}(t_1)\, p_{i_1i_2}(t_2 - t_1)\cdots p_{i_{n-1}i_n}(t_n - t_{n-1}),$$

gdzie $p_{ij}(t)$ to elementy $P(t) = e^{tQ}$.

Własności: $p_{ij}(t) = P(X_{t+s} = j \mid X_s = i)$ (jednorodność); własność Markowa (przeszłość nieistotna); rozkład w chwili $t$: $P(X_t = j) = (\nu P(t))_j$ — pełna analogia z $w = uP^n$.

## 📘 Rozkład niezmienniczy: $\pi Q = 0$ (Definicja 2.4)

$\pi$ jest **niezmienniczy**, gdy $\pi P(t) = \pi$ dla wszystkich $t$. Różniczkując po $t$ w $t=0$:

$$\pi\, Q = 0$$

i odwrotnie ($\pi Q = 0 \Rightarrow \pi Q^k = 0 \Rightarrow \pi e^{tQ} = \pi$). **Zapamiętaj:** w czasie dyskretnym warunek to $\pi P = \pi$ (wektor własny dla $\lambda = 1$), w ciągłym $\pi Q = 0$ (lewy wektor własny $Q$ dla wartości własnej $0$).

Jeśli istnieje granica $\lim_{t\to\infty}P(t) = \Pi$, to $\Pi$ ma wszystkie wiersze równe $\pi$ — rozkład stacjonarny, także niezmienniczy (Własność 4).

**📘 Przykład 2.5:** dla $Q = \begin{pmatrix} -\alpha & \alpha \\ \beta & -\beta\end{pmatrix}$ z $\alpha + \beta > 0$: $\pi = \left(\frac{\beta}{\alpha+\beta}, \frac{\alpha}{\alpha+\beta}\right)$; gdy $\alpha = \beta = 0$, $P(t) = I$ i **każdy** rozkład jest niezmienniczy.

## 📘 Teorema 2.3: mechanika ruchu — czasy pobytu i łańcuch skoków

Niech $X_t$ będzie CTMC o generatorze $Q$ i niech $q_i = -q_{ii} > 0$. Wtedy:

1. **Czas pobytu** (wł. *tempo di attesa residuo*) w stanie $i$ ma rozkład **wykładniczy** o parametrze $q_i$:
$$P(R_i \ge \tau \mid X_t = i) = e^{-q_i \tau};$$
średni czas pobytu $= 1/q_i$.
2. Po upływie tego czasu łańcuch skacze do $j \ne i$ z prawdopodobieństwem

$$\hat p_{ij} = \frac{q_{ij}}{q_i}.$$

Macierz $\hat P$ ($\hat p_{ii} = 0$) jest stochastyczna — to **łańcuch skoków** / „przyspieszony" (wł. *CM accelerata*): dyskretny łańcuch zapamiętujący tylko kolejność odwiedzanych stanów, bez czasów pobytu.

**Osservazione 2.10:** jeśli $q_i = 0$, stan $i$ jest **pochłaniający** — łańcuch zostaje w nim na zawsze.

**Osservazione 2.11 (czytanie generatora):** $Q$ to macierz **temp**: na przekątnej minus-odwrotności średnich czasów pobytu, poza przekątną tempa skoków. Dla procesu narodzin i śmierci czas pobytu w $i\ge1$ to $\min(T_1, T_2)$ (wykładnicze o parametrach $\mu_i$, $\lambda_i$) $\sim Esp(\lambda_i + \mu_i)$, a skok w górę ma prawd. $\lambda_i/(\lambda_i+\mu_i)$.

Generatory z §2.4: proces czystych narodzin — $Q$ dwudiagonalna ($-\lambda_i$, $\lambda_i$); proces Poissona — to samo z $\lambda_i = \lambda$ (a $P(t)$ ma w wierszach rozkład Poissona($\lambda t$), wzór (2.39)); narodziny i śmierci — trójdiagonalna (2.41).

**📘 Przykład 2.6 (wirus):** $N+1$ wariantów wirusa; mutacja po czasie $Esp(\lambda)$, do każdego z pozostałych wariantów równo prawdopodobnie: $q_{ij} = \lambda/N$, $q_{ii} = -\lambda$. Dla $N = 1$: $P(t)$ z Przykładu 2.3 z $\alpha=\beta=\lambda$, czyli $p_{ii}(t) = \tfrac12 + \tfrac12 e^{-2\lambda t} \to \tfrac12$.

## 📘 Esercizio 2.2 (s. 60–61 wydruku) — rozwiązany krok po kroku

Znaleźć $a$ tak, by $Q = \begin{pmatrix} a & 1/4 & 1/4 \\ 0 & -1/4 & 1/4 \\ 1/3 & 1/3 & -2/3 \end{pmatrix}$ była generatorem; potem rozkład stacjonarny i $P(0.1)$.

**1. Wartość $a$:** suma wiersza $= 0 \Rightarrow a = -1/2$.

**2. Rozkład stacjonarny:** układ $\pi Q = 0$, $\sum\pi_i = 1$:

$$\begin{cases} -\tfrac12\pi_1 + \tfrac13\pi_3 = 0 \\ \tfrac14\pi_1 - \tfrac14\pi_2 + \tfrac13\pi_3 = 0 \\ \pi_1 + \pi_2 + \pi_3 = 1\end{cases} \Rightarrow \pi = \left(\tfrac{2}{11}, \tfrac{6}{11}, \tfrac{3}{11}\right) \approx (0.18,\ 0.55,\ 0.27).$$

**3. $P(0.1)$ w przybliżeniu:** $P(h) \approx I + hQ$ dla małego $h$; dokładnie — `expm(0.1*Q)` (Scilab: `P = expm(0.1*Q)`), różnica jest nieduża.

**4. Łańcuch „przyspieszony":** $\hat p_{ij} = q_{ij}/(-q_{ii})$ daje $\hat P = \begin{pmatrix} 0 & 1/2 & 1/2 \\ 0 & 0 & 1 \\ 1/2 & 1/2 & 0\end{pmatrix}$ o rozkładzie stacjonarnym $\hat\pi = \left(\tfrac{4}{18}, \tfrac13, \tfrac49\right) \ne \pi$! Rozkłady różnią się, bo CTMC spędza w stanach **różne** czasy ($1/q_i$), a łańcuch skoków je ignoruje. (W drugiej części Esercizio 2.2: gdy wszystkie $q_i$ są równe, oba rozkłady się pokrywają.)

**📘 Esercizio 2.1:** $Q = \begin{pmatrix} -1/2 & 1/2 \\ 0 & 0\end{pmatrix}$ — stan 2 pochłaniający ($q_2 = 0$); $P(t) = \begin{pmatrix} e^{-t/2} & 1 - e^{-t/2} \\ 0 & 1 \end{pmatrix} \to \begin{pmatrix} 0&1\\0&1\end{pmatrix}$, rozkład stacjonarny $(0, 1)$.

**📘 Esercizio 2.3:** $Q$ z **dwoma** stanami pochłaniającymi ($q_{22} = q_{33} = 0$): nieskończenie wiele rozkładów niezmienniczych $(0, \theta, 1-\theta)$, rozkład stacjonarny **nie istnieje** — pełna analogia z Osservazione 1.2 dla czasu dyskretnego.

```python
import numpy as np
from fractions import Fraction
from scipy.linalg import expm

# 📘 Esercizio 2.2
Q = np.array([[-1/2, 1/4, 1/4],
              [0,   -1/4, 1/4],
              [1/3, 1/3, -2/3]])
print("sumy wierszy Q:", Q.sum(axis=1))

# rozkład stacjonarny: pi Q = 0, sum pi = 1  (układ liniowy)
A = np.vstack([Q.T, np.ones(3)])          # równania: Q^T pi^T = 0 oraz 1·pi = 1
b = np.array([0, 0, 0, 1.0])
pi, *_ = np.linalg.lstsq(A, b, rcond=None)
print("pi =", np.round(pi, 6), " = (2/11, 6/11, 3/11) =",
      (float(Fraction(2,11)), float(Fraction(6,11)), float(Fraction(3,11))))
```

```python
# P(0.1): przybliżenie I + hQ vs dokładne expm (wartości z podręcznika)
h = 0.1
print("I + 0.1*Q:\n", np.round(np.eye(3) + h*Q, 4))
print("\nexpm(0.1*Q):\n", np.round(expm(h*Q), 7))
# podręcznik: [[0.9516269, 0.0244847, 0.0238884], ...]

print("\nP(100) — wiersze = pi:")
print(np.round(expm(100*Q), 6))
```

```python
# łańcuch skoków (CM "accelerata") i jego rozkład stacjonarny
q = -np.diag(Q)
Phat = Q / q[:, None]
np.fill_diagonal(Phat, 0.0)
print("P^ (łańcuch skoków):\n", Phat)

w, v = np.linalg.eig(Phat.T)
k = np.argmin(abs(w - 1))
pihat = np.real(v[:, k]); pihat /= pihat.sum()
print("pi^ =", np.round(pihat, 6), " = (4/18, 1/3, 4/9) =",
      (4/18, 1/3, 4/9))
print("pi^ != pi — bo czasy pobytu 1/q_i =", np.round(1/q, 3), "są różne")
```

```python
# 📘 Esercizio 2.3: dwa stany pochłaniające -> brak rozkładu stacjonarnego
Q3 = np.array([[-1/2, 1/2, 0],
               [0, 0, 0],
               [0, 0, 0]])
P1 = expm(Q3)
print("P(1):\n", np.round(P1, 6))
print("\nP(1)^1000:\n", np.round(np.linalg.matrix_power(P1, 1000), 6))
# wiersz 3 zostaje (0,0,1), wiersze 1-2 dążą do (0,1,0) -> granice zależą od startu,
# więc rozkład stacjonarny nie istnieje; niezmiennicze: (0, theta, 1-theta)
```

## Typowe błędy

1. **$\pi Q = 0$, nie $\pi Q = \pi$!** Warunek niezmienniczości w czasie ciągłym dotyczy wartości własnej $0$ generatora (w dyskretnym: $1$ macierzy przejścia).
2. Rozkład stacjonarny łańcucha skoków $\hat\pi$ to na ogół **co innego** niż $\pi$ CTMC — trzeba uwzględnić czasy pobytu (chyba że wszystkie $q_i$ równe).
3. $\hat p_{ij} = q_{ij}/q_i$ wymaga $q_i > 0$; dla stanu pochłaniającego ($q_i = 0$) łańcuch skoków nie jest tam zdefiniowany (nie ma skoku).
4. $P(h) \approx I + hQ$ to tylko przybliżenie pierwszego rzędu — dla większych $t$ używaj `expm`.

## Podsumowanie

- $(\nu, Q)$-CM: rozkład początkowy $\nu$, przejścia $p_{ij}(t)$ z $P(t) = e^{tQ}$; $P(X_t = j) = (\nu P(t))_j$.
- Niezmienniczość: $\pi Q = 0 \iff \pi P(t) = \pi\ \forall t$.
- Mechanika: pobyt w $i$ przez czas $Esp(q_i)$, potem skok wg $\hat p_{ij} = q_{ij}/q_i$.
- Stany pochłaniające: $q_i = 0$; dwa takie stany $\Rightarrow$ brak rozkładu stacjonarnego.

## Odpowiedzi

1. $1/q_1 = 1/5$ jednostki czasu (czas pobytu $\sim Esp(5)$).
2. $\hat p_{12} = q_{12}/(-q_{11}) = 1/2$.
3. Warunki są równoważne (dla wszystkich $t$ naraz): $\pi Q = 0 \Rightarrow \pi e^{tQ} = \pi\ \forall t$. Ale sprawdzenie $\pi P(t_0) = \pi$ dla jednego $t_0$ wymagałoby znajomości $P(t_0) = e^{t_0Q}$ — a $\pi Q = 0$ to prosty układ liniowy na współczynnikach $Q$, bez liczenia eksponenty.
