## Czego się uczymy i po co?

Uogólniamy proces Poissona: system może nie tylko rosnąć (narodziny), ale i maleć (śmierci), a tempo zależy od aktualnego stanu. Tym modelem opisuje się dynamikę populacji, epidemie i — co najważniejsze dla nas — **kolejki** (notatnik 14 to w całości zastosowania procesów narodzin i śmierci).

## 📘 Model (s. 42 wydruku)

System w stanie $n$ (liczba całkowita) w przedziale $[t, t+h]$:
- przechodzi do $n+1$ („narodziny") z prawd. $\lambda_n h + o(h)$,
- przechodzi do $n-1$ („śmierć") z prawd. $\mu_n h + o(h)$,
- skacze o 2 lub więcej z prawd. $o(h)$,
- zostaje w $n$ z prawd. $1 - \lambda_n h - \mu_n h + o(h)$.

Współczynniki $\lambda_n \ge 0$ (**tempa narodzin**) i $\mu_n \ge 0$ (**tempa śmierci**) zależą od stanu $n$, ale nie od czasu ani historii (markowowskość).

Przypadki szczególne:
- $\mu_n = 0$: proces **czystych narodzin** (*pura nascita*);
- $\lambda_n = \lambda$, $\mu_n = 0$: proces Poissona;
- $\lambda_n = 0$: proces śmierci.

## 📘 Równania Kołmogorowa (wzór (2.10))

Tym samym rachunkiem co dla Poissona (rozbicie $\{X(t+h) = n\}$ na scenariusze $E_1,\dots,E_4$, podzielenie przez $h$, granica $h \to 0$) dostajemy dla $p_n(t) = P(X(t) = n)$:

$$\begin{cases} p_0'(t) = -\lambda_0 p_0(t) + \mu_1 p_1(t) \\ p_k'(t) = -(\lambda_k + \mu_k)\,p_k(t) + \lambda_{k-1}\,p_{k-1}(t) + \mu_{k+1}\,p_{k+1}(t), & k \ge 1 \end{cases}$$

Czytanie: zmiana $p_k$ = (dopływ z $k{-}1$ przez narodziny) + (dopływ z $k{+}1$ przez śmierć) − (odpływ z $k$).

## 📘 Rozkład stacjonarny: bilans przepływów (wzory (2.11)–(2.13))

Szukamy rozwiązań stacjonarnych $p_k(t) \equiv \pi_k$: wstawiamy $p_k' = 0$. Dostajemy **bilans globalny** (2.12):

$$(\lambda_k + \mu_k)\,\pi_k = \lambda_{k-1}\pi_{k-1} + \mu_{k+1}\pi_{k+1}$$

(w równowadze strumień prawdopodobieństwa wypływający ze stanu $k$ = strumień wpływający). Z diagramu przejść wynika mocniejszy **bilans lokalny** (2.13), dla sąsiednich par stanów:

$$\lambda_0 \pi_0 = \mu_1 \pi_1, \qquad \lambda_k \pi_k = \mu_{k+1}\pi_{k+1}.$$

Rozwiązujemy go iteracyjnie: $\pi_1 = \frac{\lambda_0}{\mu_1}\pi_0$, $\pi_2 = \frac{\lambda_1}{\mu_2}\pi_1 = \frac{\lambda_1\lambda_0}{\mu_2\mu_1}\pi_0$, … i ogólnie

$$\pi_k = \frac{\lambda_0\lambda_1\cdots\lambda_{k-1}}{\mu_1\mu_2\cdots\mu_k}\,\pi_0, \qquad \pi_0 = \left(1 + \sum_{k=1}^{\infty}\frac{\lambda_0\cdots\lambda_{k-1}}{\mu_1\cdots\mu_k}\right)^{-1}$$

(z warunku normalizacji $\sum_k \pi_k = 1$).

**Przypadek stałego ilorazu** $\lambda_{k-1}/\mu_k = \rho$ (np. $\lambda_i = \lambda$, $\mu_i = \mu$): szereg geometryczny zbiega tylko dla $\rho < 1$ i wtedy

$$\pi_k = \rho^k(1-\rho)$$

— rozkład **geometryczny** o parametrze $1-\rho$. Warunek $\rho<1$ („śmierci szybsze niż narodziny") to warunek istnienia równowagi. Jeśli $\lambda_{k-1}/\mu_k \le c < 1$, rozkład stacjonarny istnieje z kryterium porównawczego.

## 📘 Osservazione 2.3: średnia populacji przy $\lambda_n = n\lambda$, $\mu_n = n\mu$

Tempa proporcjonalne do wielkości populacji (każdy osobnik rodzi/umiera niezależnie). Rachunek warunkowy daje równanie na $M(t) = E(X(t))$:

$$M'(t) = (\lambda - \mu)M(t) \;\Longrightarrow\; M(t) = X(0)\,e^{(\lambda-\mu)t}.$$

Trzy scenariusze: $\lambda > \mu$ — wykładniczy wzrost; $\lambda = \mu$ — stała średnia; $\lambda < \mu$ — wymieranie.

```python
import numpy as np

# bilans lokalny i rozkład stacjonarny dla lambda_k = lambda, mu_k = mu
lam, mu = 2.0, 3.0
rho = lam / mu
k = np.arange(0, 20)
pi = rho**k * (1 - rho)
print("rho =", rho)
print("pi_0..pi_5 =", np.round(pi[:6], 4))
print("suma pi (20 wyrazów):", round(pi.sum(), 6), " -> reszta szeregu zaniedbywalna")

# sprawdzenie bilansu lokalnego: lambda*pi_k = mu*pi_{k+1}
print("bilans lokalny OK?", np.allclose(lam*pi[:-1], mu*pi[1:]))
```

```python
# symulacja procesu narodzin i śmierci (stałe tempa) — częstości ~ pi
rng = np.random.default_rng(seed=99)
T_max, t, stan = 50_000.0, 0.0, 0
czas_w_stanie = {}
while t < T_max:
    l_n, m_n = lam, (mu if stan > 0 else 0.0)
    czas_pobytu = rng.exponential(1 / (l_n + m_n))     # czas do najbliższego zdarzenia
    czas_w_stanie[stan] = czas_w_stanie.get(stan, 0) + czas_pobytu
    t += czas_pobytu
    stan += 1 if rng.random() < l_n / (l_n + m_n) else -1
total = sum(czas_w_stanie.values())
print("stan | ułamek czasu | pi_k")
for s in range(6):
    print(f"  {s}  |    {czas_w_stanie.get(s,0)/total:.4f}   | {pi[s]:.4f}")
```

```python
# Osservazione 2.3: średnia populacji M(t) = X0 exp((lambda-mu) t)
rng = np.random.default_rng(seed=7)
lam_i, mu_i, X0, t_kon = 1.0, 1.2, 50, 1.0   # lambda < mu -> wymieranie

def populacja_po(t_kon):
    stan, t = X0, 0.0
    while stan > 0:
        tempo = stan * (lam_i + mu_i)
        t += rng.exponential(1 / tempo)
        if t > t_kon: break
        stan += 1 if rng.random() < lam_i / (lam_i + mu_i) else -1
    return stan

wyniki = [populacja_po(t_kon) for _ in range(5_000)]
print(f"E[X(1)] symulacja: {np.mean(wyniki):.2f}")
print(f"teoria X0*exp((lam-mu)*t) = {X0*np.exp((lam_i-mu_i)*t_kon):.2f}")
```

## Typowe błędy

1. W stanie $0$ **nie ma** składnika śmierci: $\mu_0$ nie występuje (nie można zejść poniżej 0); pierwsze równanie ma inną postać.
2. Bilans lokalny (2.13) to mocniejsze stwierdzenie niż globalny (2.12) — dla procesów narodzin i śmierci wynika z „jednowymiarowej" struktury przejść (sąsiednie stany); nie zachodzi dla dowolnych łańcuchów.
3. Warunek $\rho < 1$ jest **konieczny** dla istnienia rozkładu stacjonarnego przy stałych tempach — bez niego szereg normalizujący rozbiega się i równowagi nie ma.
4. $\lambda_n, \mu_n$ to **tempa** (jednostka: 1/czas), nie prawdopodobieństwa — prawdopodobieństwem jest dopiero $\lambda_n h$ dla małego $h$.

## Podsumowanie

- Proces narodzin i śmierci: skoki $\pm1$ z tempami $\lambda_n$, $\mu_n$ zależnymi od stanu.
- Równania Kołmogorowa (2.10) opisują ewolucję $p_k(t)$; stacjonarnie $p_k' = 0$.
- Bilans lokalny $\lambda_k\pi_k = \mu_{k+1}\pi_{k+1}$ daje jawny wzór iloczynowy na $\pi_k$.
- Stałe tempa: $\pi_k = \rho^k(1-\rho)$ (geometryczny), istnieje tylko dla $\rho = \lambda/\mu < 1$.

## Odpowiedzi

1. To proces czystych narodzin ($\mu_n = 0$) ze stałym tempem $\lambda_n = \lambda$ — system tylko rośnie, skoki pojedyncze, tempo niezależne od stanu.
2. $\rho = 1/2$, więc $\pi_0 = 1 - \rho = 1/2$.
3. Nie istnieje rozkład stacjonarny: masa prawdopodobieństwa „ucieka do nieskończoności" (system rośnie bez ograniczeń — np. kolejka, w której klienci przychodzą szybciej, niż są obsługiwani).
