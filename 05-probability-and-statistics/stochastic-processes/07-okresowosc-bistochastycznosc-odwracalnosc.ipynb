## Czego się uczymy i po co?

Poznajemy głównych „winowajców" braku stacjonarności: **stany okresowe** (łańcuch oscyluje i granica nie istnieje). Do tego dwa pojęcia specjalne: macierze **bistochastyczne** (gotowy rozkład niezmienniczy: jednostajny) i **odwracalność** (równanie bilansu szczegółowego — fundament algorytmu Metropolis z notatnika 09).

## 📘 Definicja 1.5 (okresowość, s. 27 wydruku)

Stan $i$ jest **okresowy o okresie $d_i > 1$** (wł. *periodico di periodo $d_i$*), jeżeli

$$\mathrm{MCD}\{n > 0: p_{ii}^{(n)} > 0\} = d_i$$

($\mathrm{MCD}$ = NWD, największy wspólny dzielnik; po włosku *massimo comun divisore*). Jeśli NWD $=1$, stan jest **nieokresowy** (wł. *aperiodico*).

Intuicja: powroty do stanu okresowego są możliwe tylko w chwilach będących wielokrotnościami $d_i$ — łańcuch „tyka jak zegar".

Praktyczna uwaga z podręcznika: jeśli $p_{ii}^{(n)} > 0$ i $p_{ii}^{(m)} > 0$ dla $n, m$ **względnie pierwszych**, to stan nieokresowy.

**Fakt:** łańcuch ze stanami okresowymi (okres $>1$) nie może być stacjonarny — granica $\lim_n p_{ij}^{(n)}$ nie istnieje (ciąg oscyluje).

### 📘 Proposizione 1.6: okres jest własnością klasy

Jeśli $i$ i $j$ komunikują się wzajemnie, to $\mathrm{per}(i) = \mathrm{per}(j)$. (Dowód przez składanie tras $i\to j \to j \to i$ i arytmetykę podzielności.)

### Łańcuchy cykliczne

Łańcuch jest **cykliczny**, gdy wszystkie stany komunikują się i istnieje $d>0$ z $P^{d+1} = P$. Wtedy wszystkie stany mają okres $d$. Uwaga: $P^{d+1} = P$ **nie** oznacza $P^d = I$, chyba że $P$ jest nieosobliwa; może też zachodzić $P^{d+1}=P$ dla łańcucha, który nie jest cykliczny (Przykład 1.16).

## 📘 Przykłady 1.14–1.16 (s. 28–29 wydruku)

**Przykład 1.14 — cykl trójstanowy:** $P = \begin{pmatrix} 0 & 1 & 0 \\ 0 & 0 & 1 \\ 1 & 0 & 0 \end{pmatrix}$ — deterministyczna karuzela $1\to2\to3\to1$. $P^4 = P$ (oraz $P^3 = I$, bo $\det P = 1 \ne 0$); okres 3. Rozkład stacjonarny **nie istnieje** ($P^n$ nie ma granicy), ale niezmienniczy tak: $\pi = (\tfrac13,\tfrac13,\tfrac13)$ (lewy wektor własny dla $\lambda = 1$).

**Przykład 1.15 — random walk z odbiciem, 3 stany:** $P = \begin{pmatrix} 0 & 1 & 0 \\ q & 0 & p \\ 0 & 1 & 0 \end{pmatrix}$. Mamy $P^3 = P$, czyli $P^{2n}=P^2$, $P^{2n+1}=P$; wszystkie stany mają okres 2 ($p^{(2n+1)}_{ii}=0$). Stacjonarnego brak; niezmienniczy: $\pi = (q/2,\ 1/2,\ p/2)$.

**Przykład 1.16 — okresowość bez cykliczności:** $P = \begin{pmatrix} 0 & 1 & 0 \\ 0 & 0 & 1 \\ 0 & 1 & 0 \end{pmatrix}$. Stan 1 chwilowy, stany 2, 3 powracające o okresie 2. $P^3 = P$, ale łańcuch nie jest cykliczny (nie wszystkie stany w jednej klasie). Niezmienniczy: $\pi = (0, \tfrac12, \tfrac12)$.

## 📘 Definicja 1.6 (macierz bistochastyczna) i Przykład 1.17

Macierz stochastyczna $P$ jest **bistochastyczna** (wł. *bistocastica*, podwójnie stochastyczna), jeśli także **kolumny** sumują się do 1: $\sum_{i} p_{ij} = 1$ (równoważnie: $P^T$ też jest stochastyczna).

**Fakt:** dla bistochastycznej $P$ rozkład **jednostajny** $v = (\tfrac1N, \dots, \tfrac1N)$ jest niezmienniczy (sprawdzenie: $\sum_i \tfrac1N p_{ij} = \tfrac1N \sum_i p_{ij} = \tfrac1N$). Uwaga: nie musi być jedyny ani stacjonarny!

**Przykład 1.17:** $P$ blokowa $\begin{pmatrix} P_1 & 0 \\ 0 & P_2 \end{pmatrix}$ z $P_1 = \begin{pmatrix} 0&1\\1&0\end{pmatrix}$ (okres 2), $P_2 = \begin{pmatrix} 1/2&1/2\\1/2&1/2\end{pmatrix}$ (nieokresowe). Dwie klasy zamknięte $\Rightarrow$ stacjonarnego brak; bistochastyczność $\Rightarrow$ niezmienniczy $\pi = (\tfrac14,\tfrac14,\tfrac14,\tfrac14)$.

## 📘 Definicja 1.7 (odwracalność i bilans szczegółowy, s. 30 wydruku)

Rozkład $\pi$ jest **odwracalny** (wł. *reversibile*), jeżeli spełnia **równanie bilansu szczegółowego** (wł. *equazione del bilancio dettagliato*):

$$\pi_i\, p_{ij} = \pi_j\, p_{ji} \qquad \forall i, j \in E. \tag{1.40}$$

**Fakt:** odwracalny $\Rightarrow$ niezmienniczy. Dowód (jedna linijka): $\sum_i \pi_i p_{ij} = \sum_i \pi_j p_{ji} = \pi_j \sum_i p_{ji} = \pi_j$. ✔

**Interpretacja:** w równowadze „strumień prawdopodobieństwa" $i \to j$ (czyli $\pi_i p_{ij}$) równa się strumieniowi $j \to i$. Łańcuch oglądany „od tyłu" (sekwencja $X_n, X_{n-1}, \dots$) ma wtedy te same prawdopodobieństwa przejścia co oryginalny — stąd nazwa. Podręcznik wyprowadza: łańcuch odwrócony ma $q_{ij} \approx \pi_j p_{ji}/\pi_i$, a odwracalność to $q_{ij} = p_{ij}$.

Bilans szczegółowy to **mocniejszy** warunek niż niezmienniczość ($\pi P = \pi$ to „bilans globalny": całkowity dopływ = odpływowi). Jest za to dużo łatwiejszy do sprawdzenia — i na nim opiera się algorytm Metropolis (notatnik 09).

```python
import numpy as np
from fractions import Fraction

# 📘 Przykład 1.14: cykl — brak granicy P^n, ale istnieje rozkład niezmienniczy
P = np.array([[0, 1, 0],
              [0, 0, 1],
              [1, 0, 0]], dtype=float)
for n in [1, 2, 3, 4, 5, 6]:
    print(f"P^{n} =\n{np.linalg.matrix_power(P, n).astype(int)}")
# P^3 = I, P^4 = P: wartości p_ii^(n) oscylują -> granica nie istnieje
```

```python
# niezmienniczość rozkładu jednostajnego (P jest bistochastyczna):
pi = np.array([1/3, 1/3, 1/3])
print("pi P =", pi @ P, " == pi:", np.allclose(pi @ P, pi))

# 📘 Przykład 1.15: random walk z odbiciem, p=0.3
p = 0.3; q = 1 - p
P15 = np.array([[0, 1, 0],
                [q, 0, p],
                [0, 1, 0]])
pi15 = np.array([q/2, 1/2, p/2])
print("\nPrzykład 1.15: pi P =", pi15 @ P15, " == pi:", np.allclose(pi15 @ P15, pi15))
print("P^2 == P^4:", np.allclose(np.linalg.matrix_power(P15,2), np.linalg.matrix_power(P15,4)),
      "-> okres 2, granica P^n nie istnieje")
```

```python
# 🧪 Przykład własny: sprawdzenie bilansu szczegółowego
# Random walk z Przykładu 1.15 i jego pi spełniają bilans szczegółowy:
def bilans_szczegolowy(pi, P, atol=1e-12):
    N = len(pi)
    return all(abs(pi[i]*P[i,j] - pi[j]*P[j,i]) < atol for i in range(N) for j in range(N))

print("RW z odbiciem (1.15): bilans szczegółowy?", bilans_szczegolowy(pi15, P15))

# ...a cykl z Przykładu 1.14 NIE jest odwracalny (strumień krąży w jedną stronę):
print("cykl (1.14):          bilans szczegółowy?", bilans_szczegolowy(pi, P))
# pi_1 p_12 = 1/3 * 1 = 1/3, ale pi_2 p_21 = 1/3 * 0 = 0 — brak równowagi par
```

## Typowe błędy

1. Okres to **NWD** wszystkich możliwych czasów powrotu, nie najkrótszy czas powrotu. Stan z $p_{ii}^{(2)}>0$ i $p_{ii}^{(3)}>0$ ma okres 1 (NWD(2,3)=1), mimo że nie wraca w 1 kroku.
2. $P^{d+1} = P$ nie implikuje $P^d = I$ (tylko dla nieosobliwej $P$) ani cykliczności (Przykład 1.16).
3. Bistochastyczność daje **niezmienniczość** rozkładu jednostajnego — nie stacjonarność i nie jedyność (Przykład 1.17 ma niezmienniczy jednostajny, a stacjonarnego brak).
4. Niezmienniczość NIE implikuje odwracalności — cykl 1.14 ma $\pi$ niezmienniczy, ale strumienie krążą jednokierunkowo (bilans globalny OK, szczegółowy nie).

## Podsumowanie

- Okres stanu: $d_i = \mathrm{NWD}\{n: p_{ii}^{(n)} > 0\}$; okres $>1$ wyklucza stacjonarność; okres jest stały w klasie.
- Bistochastyczna: wiersze i kolumny sumują się do 1 $\Rightarrow$ jednostajny rozkład niezmienniczy.
- Odwracalność: $\pi_i p_{ij} = \pi_j p_{ji}$ (bilans szczegółowy) $\Rightarrow$ niezmienniczość; silniejsze i łatwiejsze do sprawdzenia.

## Mini słowniczek IT–PL

| włoski | polski |
|---|---|
| stato periodico / aperiodico | stan okresowy / nieokresowy |
| periodo | okres |
| MCD (massimo comun divisore) | NWD (największy wspólny dzielnik) |
| catena ciclica | łańcuch cykliczny |
| matrice bistocastica | macierz bistochastyczna (podwójnie stochastyczna) |
| distribuzione uniforme | rozkład jednostajny |
| (distribuzione) reversibile | (rozkład) odwracalny |
