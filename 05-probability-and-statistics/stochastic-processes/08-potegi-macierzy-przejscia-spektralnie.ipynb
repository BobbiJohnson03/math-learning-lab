## Czego się uczymy i po co?

Jak **jawnie policzyć** $P^n$ bez mnożenia macierzy $n$ razy? Odpowiedź: przez wartości własne (diagonalizację). Dodatkowy zysk: widmo mówi nam, **jak szybko** łańcuch zbiega do równowagi — rządzi tym drugi co do wielkości moduł wartości własnej.

## 📘 Widmo macierzy stochastycznej (s. 31 wydruku)

**Fakt:** każda macierz stochastyczna $P$ ma wartość własną $\lambda_1 = 1$ (prawy wektor własny: kolumna jedynek), a wszystkie wartości własne spełniają $|\lambda| \le 1$ (leżą w kole jednostkowym na płaszczyźnie zespolonej).

Dowód ograniczenia (z podręcznika): niech $Pu = \lambda u$, wybierz $h$ z $|u_h| = \max_j |u_j|$. Wtedy

$$|\lambda|\,|u_h| = \Big|\sum_j p_{hj} u_j\Big| \le \sum_j p_{hj}|u_j| \le |u_h| \sum_j p_{hj} = |u_h|,$$

i po skróceniu $|u_h| > 0$ dostajemy $|\lambda| \le 1$. (Tu $u$ jest **prawym, kolumnowym** wektorem własnym.)

## 📘 Diagonalizacja i $P^n$

Jeśli $P$ ma $N$ **różnych** wartości własnych, to jest diagonalizowalna: $P = U\Lambda U^{-1}$, gdzie $\Lambda = \mathrm{diag}(\lambda_1,\dots,\lambda_N)$, a kolumny $U$ to niezależne prawe wektory własne. Wtedy

$$P^n = U\,\Lambda^n\, U^{-1}, \qquad \Lambda^n = \mathrm{diag}(\lambda_1^n, \dots, \lambda_N^n)$$

— potęgowanie macierzy diagonalnej to potęgowanie liczb na przekątnej. Cała trudność $P^n$ znika!

## 📘 Jawny wzór dla dwóch stanów (wzory (1.42)–(1.47))

Niech $P = \begin{pmatrix} 1-p & p \\ a & 1-a \end{pmatrix}$, $a, p \in (0,1)$ (regularny łańcuch 2-stanowy). Wartości własne: $\lambda_1 = 1$, $\lambda_2 = 1 - a - p$ ($|\lambda_2| < 1$). Prawe wektory własne $v_1 = \binom{1}{1}$, $v_2 = \binom{1}{-a/p}$. Po przeliczeniu $U \Lambda^n U^{-1}$:

$$P^n = \begin{pmatrix} \frac{a}{a+p} & \frac{p}{a+p} \\ \frac{a}{a+p} & \frac{p}{a+p} \end{pmatrix} + \frac{(1-a-p)^n}{a+p}\begin{pmatrix} p & -p \\ -a & a \end{pmatrix}. \tag{1.44}$$

Czytanie wzoru:
- pierwsza macierz ma **równe wiersze** $\pi = \left(\frac{a}{a+p}, \frac{p}{a+p}\right)$ — to rozkład stacjonarny (lewy wektor własny dla $\lambda=1$);
- druga część gaśnie jak $\lambda_2^n$ — **im $|\lambda_2|$ bliższe zera, tym szybsza zbieżność do równowagi**;
- kompaktowo: $P^n = \Pi + \lambda_2^n R$, gdzie $\Pi$ — macierz o wierszach $\pi$, $R$ — stała macierz (niestochastyczna).

## 📘 Teorema 1.2 (ergodico II) i Osservazione 1.5

**Teorema 1.2:** łańcuch o skończonym $E$, **nieprzywiedlny** i o wszystkich stanach **nieokresowych** ma rozkład stacjonarny. (Działa też, gdy łańcuch nie jest regularny.)

**Osservazione 1.5 (widmo a regularność):** jeśli $P$ regularna, to $1$ jest **pojedynczą** (prostą) wartością własną, a pozostałe mają moduł $<1$. **Odwrotnie — NIE:** podręcznik podaje macierz

$$P = \begin{pmatrix} 1 & 0 & 0 \\ 1/2 & 1/4 & 1/4 \\ 0 & 1/2 & 1/2 \end{pmatrix}$$

o widmie $\{1,\ 3/4,\ 0\}$ (spełnia warunki widmowe), która regularna nie jest (każda potęga ma zera — stan 1 jest pochłaniający). Rozkład stacjonarny istnieje: $\pi = (1, 0, 0)$. **Morał: nie wolno wnioskować regularności z widma ani z wyznacznika.**

## 📘 Metoda współczynników dla 3 stanów (s. 33–35 wydruku)

Dla $P$ diagonalizowalnej o wartościach własnych $1, \lambda_2, \lambda_3$ (różnych) każdy element potęgi ma postać:

$$p_{ij}^{(n)} = A_{ij} + B_{ij}\,\lambda_2^n + C_{ij}\,\lambda_3^n$$

⚠️ **$A_{ij}, B_{ij}, C_{ij}$ to indeksowane WSPÓŁCZYNNIKI (stałe liczby zależne od pary $i,j$), a NIE potęgi!** Wyznacza się je z $\Gamma$ i $\Gamma^{-1}$ ($A_{ij} = \Gamma_{i1}(\Gamma^{-1})_{1j}$ itd.) — albo sprytniej, z małych $n$:

### 📘 Przykład z podręcznika: $P = \begin{pmatrix} 1/3 & 0 & 2/3 \\ 1/3 & 2/3 & 0 \\ 1/3 & 1/3 & 1/3 \end{pmatrix}$, widmo $1, \tfrac13, 0$

Dla $n \ge 1$: $p_{ij}^{(n)} = A_{ij} + B_{ij}(\tfrac13)^n$ (człon $C_{ij}\cdot 0^n$ znika dla $n\ge 1$).

Liczymy $p_{11}^{(n)}$: znamy $p_{11} = \tfrac13$ oraz (z Chapman–Kołmogorowa) $p_{11}^{(2)} = \tfrac19 + 0 + \tfrac23\cdot\tfrac13 = \tfrac13$. Układ:

$$\begin{cases} A + B/3 = 1/3 \\ A + B/9 = 1/3 \end{cases} \Rightarrow A = \tfrac13,\ B = 0 \Rightarrow p_{11}^{(n)} = \tfrac13.$$

Dla $p_{12}^{(n)}$: $p_{12} = 0$, $p_{12}^{(2)} = \tfrac29$, układ $A + B/3 = 0$, $A + B/9 = \tfrac29$ daje $A = \tfrac13$, $B = -1$, czyli $p_{12}^{(n)} = \tfrac13 - (\tfrac13)^n$. Z dopełnienia: $p_{13}^{(n)} = \tfrac13 + (\tfrac13)^n$ *(drobne potknięcie rachunkowe w źródle przy pośrednim kroku nie zmienia wyniku)*.

Granica: $p_{1j}^{(n)} \to \tfrac13$ — zgodnie z twierdzeniem ergodycznym ($P$ regularna, bo $P^2 > 0$; bistochastyczna, więc $\pi$ jednostajny).

### 📘 Osservazione 1.6

Jeśli istnieją **dwa różne** rozkłady niezmiennicze, to rozkład stacjonarny nie istnieje. (Gdyby istniał $\pi$, to każdy niezmienniczy $\mu$ spełniałby $\mu = \mu\Pi = \pi$.)

```python
import numpy as np

# 📘 weryfikacja wzoru (1.44) dla 2 stanów
a_, p_ = 0.3, 0.2
P = np.array([[1-p_, p_], [a_, 1-a_]])
lam2 = 1 - a_ - p_
Pi = np.array([[a_/(a_+p_), p_/(a_+p_)], [a_/(a_+p_), p_/(a_+p_)]])
R = np.array([[p_, -p_], [-a_, a_]]) / (a_ + p_)

for n in [1, 2, 5, 20]:
    wzor = Pi + lam2**n * R
    potega = np.linalg.matrix_power(P, n)
    print(f"n={n:2d}: max|wzór - matrix_power| = {abs(wzor - potega).max():.2e}")
```

```python
# 📘 przykład 3-stanowy: p_ij^(n) = A + B (1/3)^n
P3 = np.array([[1/3, 0,   2/3],
               [1/3, 2/3, 0  ],
               [1/3, 1/3, 1/3]])
print("wartości własne:", np.round(np.linalg.eigvals(P3), 6))

for n in [1, 2, 3, 5, 10]:
    Pn = np.linalg.matrix_power(P3, n)
    teoria_12 = 1/3 - (1/3)**n
    print(f"n={n:2d}: p_12^(n) = {Pn[0,1]:.6f}, teoria 1/3-(1/3)^n = {teoria_12:.6f}")
```

```python
# 📘 Osservazione 1.5: widmo OK, ale macierz nieregularna
P5 = np.array([[1,   0,    0  ],
               [1/2, 1/4, 1/4],
               [0,   1/2, 1/2]])
print("widmo:", np.round(np.linalg.eigvals(P5), 4), " (1 proste, reszta <1 co do modułu)")
Pn = P5.copy(); regularna = False
for n in range(1, 50):
    if (Pn > 0).all(): regularna = True; break
    Pn = Pn @ P5
print("regularna?", regularna, " (stan 1 pochłaniający -> kolumny 2,3 wiersza 1 zawsze 0)")
print("P^100:\n", np.round(np.linalg.matrix_power(P5, 100), 6), "\n-> rozkład stacjonarny (1,0,0)")
```

## Typowe błędy

1. **$A_{ij}^{(k)}$-style współczynniki to nie potęgi** — w zapisie $p_{ij}^{(n)} = A_{ij} + B_{ij}\lambda_2^n$ tylko $\lambda_2^n$ jest potęgą; $A_{ij}, B_{ij}$ to stałe indeksowane parą $(i,j)$.
2. Wnioskowanie regularności z widma/wyznacznika — kontrprzykład wyżej. Widmo daje warunek **konieczny**, nie dostateczny.
3. Metoda wymaga **diagonalizowalności** — przy wielokrotnych wartościach własnych o niepełnej krotności geometrycznej pojawiają się człony $n\lambda^n$ (przykład w notatniku 12 przy $Q$-macierzach).
4. Numerycznie: do $P^n$ używaj `matrix_power`, a wartości własnych tylko do analizy — ręczna diagonalizacja w arytmetyce zmiennoprzecinkowej bywa niestabilna.

## Podsumowanie

- Macierz stochastyczna: $\lambda_1 = 1$ zawsze, reszta widma w kole jednostkowym.
- $P = U\Lambda U^{-1} \Rightarrow P^n = U\Lambda^n U^{-1}$; elementy: kombinacje $\lambda_h^n$ ze stałymi współczynnikami.
- 2 stany: $P^n = \Pi + \lambda_2^n R$; szybkość zbieżności rządzona przez $|\lambda_2|$.
- Ergodico II: nieprzywiedlny + nieokresowy (skończony) $\Rightarrow$ stacjonarny istnieje.

## Odpowiedzi

1. W notacji wzoru: $p = 0.5$, $a = 0.3$, więc $\lambda_2 = 1 - 0.8 = 0.2$, $\pi = \left(\frac{0.3}{0.8}, \frac{0.5}{0.8}\right) = (0.375,\ 0.625)$.
2. Dla $n\ge1$: $0^n = 0$. Dla $n=0$: $P^0 = I$, a $0^0 = 1$ w tej konwencji — człon $C_{ij}$ jest potrzebny, żeby odtworzyć macierz jednostkową.
3. Ten z $\lambda_2 = 0.1$: błąd $|p_{ij}^{(n)} - \pi_j|$ maleje jak $0.1^n$ — każdy krok redukuje odchylenie 10-krotnie, zamiast o 10%.
