## Czego się uczymy i po co?

Pytanie kluczowe: **co robi łańcuch po bardzo długim czasie?** Czy prawdopodobieństwa $p_{ij}^{(n)}$ stabilizują się? Czy zapominają, skąd wystartowaliśmy? Te pytania prowadzą do najważniejszego twierdzenia rozdziału — twierdzenia ergodycznego.

## ⚠️ Terminologia Abundo: „stazionaria" ≠ „invariante"!

To **najważniejsza pułapka terminologiczna** całego podręcznika. Abundo rozróżnia:

- **Rozkład niezmienniczy** (*distribuzione invariante*, Def. 1.3): $v$ takie, że $v = vP$. Czysto algebraiczny warunek.
- **Rozkład stacjonarny** (*distribuzione stazionaria*, §1.6): $\pi$ takie, że istnieje granica

$$\pi_j = \lim_{n\to\infty} p_{ij}^{(n)} \quad\text{dla każdego } j, \textbf{ niezależnie od stanu startowego } i. \tag{1.34}$$

Związek (z przejścia granicznego w Chapman–Kołmogorowie, wzór (1.35)–(1.36)): **jeśli rozkład stacjonarny istnieje, to jest niezmienniczy** ($\pi = \pi P$) i jest jedyny. Ale **nie odwrotnie**: rozkład może być niezmienniczy, choć granica $\lim_n p_{ij}^{(n)}$ w ogóle nie istnieje (np. łańcuchy okresowe — notatnik 07).

**Uwaga przy czytaniu innych źródeł:** w większości literatury (również polskiej i angielskiej) „stationary distribution" = nasz *invariante* ($\pi P = \pi$), a to, co Abundo nazywa stacjonarnym, nosi nazwę **rozkładu granicznego** (limiting distribution). Czytając podręcznik Abundo, trzymaj się jego konwencji; czytając cokolwiek innego — sprawdź definicję!

Zapis $\pi = \pi P$ oznacza, że $\pi$ jest **lewym wektorem własnym** $P$ dla wartości własnej $1$ (równoważnie: $\pi^T$ jest prawym, kolumnowym wektorem własnym macierzy transponowanej $P^T$).

## 📘 Ergodyczność i częstości wizyt (Proposizione 1.3)

System nazywamy **ergodycznym**, gdy średnia po czasie = średnia po stanach (w równowadze):

$$\lim_{M\to\infty} \frac{1}{M}\sum_{n=1}^{M} f(X_n) = \sum_{i=1}^N f(i)\,\pi_i.$$

Biorąc $f = \mathbb{1}_{\{i\}}$ (wskaźnik stanu $i$): **częstość wizyt w stanie $i$ dąży do $\pi_i$**.

**Proposizione 1.3.** Jeżeli istnieje rozkład stacjonarny $\pi$ ($\lim_n p_{ij}^{(n)} = \pi_j$), to częstość wizyt w stanie $j$ zbiega **według prawdopodobieństwa** do $\pi_j$. Dowód (szkic z podręcznika): średnia $\tfrac1n\sum_k Y_k$ (gdzie $Y_k$ — wskaźnik $X_k = j$) ma średnią dążącą do $\pi_j$ (zbieżność średnich Cesàro) i wariancję dążącą do 0; teza z nierówności Czebyszewa.

Dlatego łańcuch posiadający rozkład stacjonarny nazywa się też **ergodycznym**.

## 📘 Definicja 1.4 (regularność) i kryterium

Łańcuch jest **regularny** (wł. *regolare*), jeżeli istnieją liczba całkowita $s > 0$ i $\alpha \in (0,1)$ takie, że

$$p_{ij}^{(s)} \ge \alpha \quad \forall i, j \in E,$$

czyli: **pewna potęga $P^s$ ma wszystkie elementy ściśle dodatnie**.

**Kryterium (s. 23):** jeśli łańcuch jest nieprzywiedlny i istnieje $h$ z $p_{hh} > 0$ (choć jeden dodatni element na przekątnej), to łańcuch jest regularny. Idea dowodu: pętla w $h$ pozwala „dopasować długości" tras $i \to h \to\dots\to h \to j$, tak by wszystkie pary $(i,j)$ były osiągalne w tej samej liczbie kroków $2m$.

⚠️ **Czego NIE wolno:** wnioskować regularności z $\det P \ne 0$ albo z samych wartości własnych. Podręcznik (Osservazione 1.5, s. 32–33) podaje wprost kontrprzykład: macierz z $1$ jako prostą wartością własną i resztą widma w kole jednostkowym, która NIE jest regularna. Regularność sprawdzamy badając **dodatniość elementów potęg $P^s$**.

## 📘 Teorema 1.1 (ergodico I, s. 23 wydruku)

Niech $X_k$ będzie łańcuchem **regularnym** na $E = \{1,\dots,N\}$ (czyli $p_{ij}^{(s)} \ge \alpha > 0$). Wtedy:

1. istnieje **dokładnie jeden** rozkład niezmienniczy $\pi$ ($\pi P = \pi$), przy czym $\pi_i > 0$ dla wszystkich $i$;
2. dla każdego $j$ istnieje granica $\lim_{n\to\infty} p_{ij}^{(n)} = \pi_j$, **niezależna od $i$** (więc $\pi$ jest też rozkładem stacjonarnym);
3. zbieżność jest wykładniczo szybka:

$$|p_{ij}^{(n)} - \pi_j| \le 2\beta^{\,n-s}, \qquad \beta = (1-\alpha)^{1/s} < 1. \tag{1.38}$$

**Obrazowo:** dla regularnego łańcucha $P^n$ dąży do macierzy, której **wszystkie wiersze są równe** wektorowi $\pi$ — skądkolwiek startujesz, po długim czasie rozkład jest ten sam.

### 📘 Obserwacje o nieistnieniu rozkładu stacjonarnego (s. 25)

- *Osservazione 1.2:* dwa stany pochłaniające $\Rightarrow$ rozkład stacjonarny nie istnieje (granice $\lim p_{12}^{(n)} = 0$ i $\lim p_{22}^{(n)} = 1$ dałyby sprzeczne $\pi_2$).
- *Osservazione 1.3:* dwie rozłączne klasy zamknięte (z których jedna regularna) $\Rightarrow$ rozkład stacjonarny nie istnieje.
- *Osservazione 1.4 + Proposizione 1.4:* łańcuch **nieprzywiedlny** (nawet nieregularny) ma zawsze **dokładnie jeden rozkład niezmienniczy** — ale niekoniecznie stacjonarny. Dowód używa sprytnej macierzy $Q = \sum_n \frac{1}{2^{n+1}}P^n$, która jest regularna, a ma te same rozkłady niezmiennicze co $P$.

## 📘 Przykład 1.11 (s. 23–24): stacjonarność bez regularności

$$P = \begin{pmatrix} 1/3 & 2/3 & 0 & 0 \\ 3/4 & 1/8 & 1/8 & 0 \\ 0 & 0 & 1/2 & 1/2 \\ 0 & 0 & 1/3 & 2/3 \end{pmatrix}$$

Każda potęga $P^n$ zawiera zera (ze stanów $3, 4$ nie wrócimy do $1, 2$), więc łańcuch **nie jest regularny** i Teorema 1.1 nie stosuje się. Mimo to **rozkład stacjonarny istnieje**: numerycznie $P^n$ zbiega do macierzy o wszystkich wierszach $\pi = (0,\, 0,\, \tfrac25,\, \tfrac35)$. Komponenty nie są wszystkie dodatnie — to możliwe właśnie dlatego, że łańcuch nie jest regularny. (Stany 1, 2 są chwilowe — po długim czasie cała masa prawdopodobieństwa przenosi się do klasy $\{3,4\}$.)

## 📘 Czas powrotu (Proposizione 1.5, s. 26)

Niech $T_i = \inf\{n \ge 1: X_n = i\}$ i $m_i = E(T_i \mid X_0 = i)$ — **średni czas pierwszego powrotu** do $i$. Dla łańcucha ergodycznego:

$$m_i = \frac{1}{\pi_i}.$$

Intuicja: jeśli łańcuch spędza w stanie $i$ ułamek czasu $\pi_i$, to między wizytami mija średnio $1/\pi_i$ kroków.

**📘 Przykład 1.12 (kostka):** rzuty kostką jako łańcuch z $p_{ij} = 1/6$; $\pi = (\tfrac16,\dots,\tfrac16)$, więc średni czas między dwoma wypadnięciami tej samej ścianki to $1/\tfrac16 = 6$. ✔ (zgadza się z rozkładem geometrycznym)

**📘 Przykład 1.13 (estymacja parametru):** łańcuch $P = \begin{pmatrix} 1/2 & 1/2 \\ p & 1-p \end{pmatrix}$ obserwowany długo spędza ~20% czasu w stanie 1 i ~80% w stanie 2. Przyjmując $\pi = (0.2, 0.8)$ i rozwiązując $\pi = \pi P$: $0.2 = 0.1 + 0.8p \Rightarrow p = 1/8$.

````python
import numpy as np

# 📘 Przykład 1.11: brak regularności, ale istnieje rozkład stacjonarny
P = np.array([[1/3, 2/3, 0,   0  ],
              [3/4, 1/8, 1/8, 0  ],
              [0,   0,   1/2, 1/2],
              [0,   0,   1/3, 2/3]])

# regularność: czy jakaś potęga ma wszystkie elementy > 0?
Pn = P.copy()
for n in range(1, 30):
    if (Pn > 0).all():
        print(f"P^{n} > 0 — regularny"); break
    Pn = Pn @ P
else:
    print("do n=30 każda potęga zawiera zera -> łańcuch NIE jest regularny")

P100 = np.linalg.matrix_power(P, 100)
print("\nP^100 (wiersze prawie równe):")
print(np.round(P100, 6))
print("\npi = (0, 0, 2/5, 3/5) =", (0, 0, 2/5, 3/5))
````

````python
# 📘 Przykład 1.13: estymacja p z rozkładu stacjonarnego
# pi = (0.2, 0.8), warunek pi = pi P daje 0.2 = 0.2*0.5 + 0.8*p
p = (0.2 - 0.1) / 0.8
print("oszacowane p =", p, "= 1/8")

# sprawdzenie: łańcuch z tym p faktycznie spędza 20% czasu w stanie 1
P2 = np.array([[0.5, 0.5], [p, 1 - p]])
rng = np.random.default_rng(seed=123)
stan, licznik, N = 0, 0, 200_000
for _ in range(N):
    stan = rng.choice(2, p=P2[stan])
    licznik += (stan == 0)
print(f"częstość stanu 1 w symulacji: {licznik/N:.4f}  (teoria: 0.2)")

# średni czas powrotu m_i = 1/pi_i (Proposizione 1.5)
print("m_1 = 1/0.2 =", 1/0.2, ",  m_2 = 1/0.8 =", 1/0.8)
````

## Typowe błędy

1. **Mylenie „invariante" ze „stazionaria"** — u Abundo stacjonarność to istnienie granicy $p_{ij}^{(n)}$ niezależnej od $i$; niezmienniczość to tylko $\pi P = \pi$. Stacjonarny $\Rightarrow$ niezmienniczy, ale nie odwrotnie!
2. **Wnioskowanie regularności ze spektrum lub wyznacznika** — zabronione; kontrprzykład w Osservazione 1.5. Regularność = dodatniość elementów pewnej potęgi.
3. Granica $\pi_j = \lim p_{ij}^{(n)}$ musi być **wspólna dla wszystkich $i$** — samo istnienie granic zależnych od $i$ (jak przy dwóch stanach pochłaniających) to za mało.
4. $m_i = 1/\pi_i$ wymaga ergodyczności — nie stosuj do łańcuchów okresowych czy przywiedlnych.

## Podsumowanie

- Stacjonarny (Abundo): $\pi_j = \lim_n p_{ij}^{(n)}$ niezależnie od $i$; wtedy $\pi=\pi P$ i $P^n \to$ macierz o równych wierszach.
- Regularny: $P^s > 0$ dla pewnego $s$; kryterium: nieprzywiedlny + dodatni element na przekątnej.
- Teorema 1.1: regularny $\Rightarrow$ istnieje jedyny $\pi > 0$, stacjonarny, zbieżność wykładnicza.
- Nieprzywiedlny $\Rightarrow$ jedyny rozkład niezmienniczy (Prop. 1.4), ale stacjonarność może zawieść.
- Ergodyczny: częstości wizyt $\to \pi_i$; średni czas powrotu $m_i = 1/\pi_i$.
