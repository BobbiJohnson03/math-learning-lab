## Czego się uczymy i po co?

To pierwszy **duży rachunek** na łańcuchu Markowa: liczymy prawdopodobieństwo, że gracz straci cały kapitał, oraz średni czas trwania gry. Przy okazji poznajemy **równania różnicowe** (wł. *equazioni alle differenze*) — dyskretny odpowiednik równań różniczkowych, który wróci przy problemach pochłaniania (notatnik 05).

## 📘 Model (Przykład 1.8, s. 8 wydruku)

Dwóch graczy. W każdej partii pierwszy gracz wygrywa 1 euro z prawd. $p \in (0,1)$ albo przegrywa 1 euro z prawd. $q = 1-p$ (nie ma remisów). Kapitał łączny jest stały i wynosi $a$; pierwszy gracz startuje z kapitałem $i$ ($0 \le i \le a$). Gra kończy się, gdy ktoś ma $0$ (jest **zrujnowany**) albo całe $a$.

To dokładnie **błądzenie losowe z barierami pochłaniającymi** $0$ i $a$: pozycja cząstki = kapitał pierwszego gracza.

Szukamy: $q_i$ = prawdopodobieństwo ruiny (pochłonięcia w $0$) przy starcie z $i$.

### Równanie różnicowe (wzór (1.8))

$$q_i = p\, q_{i+1} + q\, q_{i-1}, \qquad q_0 = 1,\quad q_a = 0.$$

**Skąd to?** Z pierwszego kroku (wzór na prawdopodobieństwo całkowite): startując z $i$, z prawd. $p$ idziemy do $i+1$ i stamtąd ruina ma prawd. $q_{i+1}$; z prawd. $q$ idziemy do $i-1$ i ruina ma prawd. $q_{i-1}$. Warunki brzegowe: start z $0$ = ruina pewna ($q_0=1$), start z $a$ = ruina niemożliwa ($q_a=0$).

Zwróć uwagę na niewygodę notacyjną: $q = 1-p$ to liczba, a $q_i$ to szukane prawdopodobieństwa — indeks robi różnicę!

## 📘 Rozwiązanie krok po kroku

### Przypadek 1: $p \ne q$

Szukamy rozwiązań postaci $q_i = \alpha^i$ (jak $e^{\lambda x}$ dla równań różniczkowych). Podstawiamy:

$$\alpha^i = p\,\alpha^{i+1} + q\,\alpha^{i-1} \;\Longrightarrow\; p\alpha^2 - \alpha + q = 0.$$

Równanie kwadratowe ma wyróżnik $1 - 4pq > 0$ (bo $p\ne q$) i pierwiastki $\alpha_1 = 1$, $\alpha_2 = q/p$. Rozwiązanie ogólne to kombinacja liniowa:

$$q_i = A + B\left(\frac{q}{p}\right)^{\!i}.$$

Z warunków brzegowych $q_0 = 1$, $q_a = 0$ wyznaczamy $A, B$ i dostajemy (wzór (1.11)):

$$q_i = \frac{(q/p)^a - (q/p)^i}{(q/p)^a - 1}.$$

### Przypadek 2: $p = q = \tfrac12$ (gra sprawiedliwa)

Pierwiastek podwójny $\alpha = 1$; rozwiązania fundamentalne to $1$ oraz $i$, więc $q_i = A + B\,i$. Z warunków brzegowych (wzór (1.12)):

$$q_i = 1 - \frac{i}{a}.$$

### Wnioski z podręcznika

- $p_i = 1 - q_i$ (prawd. wygrania wszystkiego) spełnia $p_i = q_{a-i}$ — wygrana gracza to ruina przeciwnika.
- Średni zysk $G$: $E(G) = a(1-q_i) - i$ (wzór (1.13)); dla $p = \tfrac12$ wychodzi $E(G) = 0$ — **gra sprawiedliwa** (wł. *gioco equo*).
- Gra z nieskończenie bogatym bankiem ($a \to \infty$, wzór (1.14)): $\lim_{a\to\infty} q_i = 1$ gdy $p \le q$, a $(q/p)^i$ gdy $p > q$. Nawet przy grze sprawiedliwej **ruina jest pewna**!

## 📘 Średni czas trwania gry (durata media del gioco)

Niech $\tau_i$ = losowy czas trwania gry przy kapitale początkowym $i$, a $D_i = E(\tau_i)$. Podręcznik podaje (wzór (1.15)):

$$D_i = p\, D_{i+1} + q\, D_{i-1} + 1, \qquad D_0 = D_a = 0.$$

To samo rozumowanie pierwszego kroku, ale z „$+1$", bo wykonany krok kosztuje jedną partię. Rozwiązania:

$$D_i = \frac{i}{q-p} - \frac{a}{q-p}\cdot\frac{1-(q/p)^i}{1-(q/p)^a} \quad (p \ne q), \qquad\qquad D_i = i\,(a-i) \quad (p = q = \tfrac12).$$

Dla gry sprawiedliwej: start ze środka ($i = a/2$) daje $D_{a/2} = a^2/4$ — zaskakująco długo! Dla $a=100$ średnio 2500 partii.

## Przykład liczbowy z pełnym rachunkiem (🧪 własny, weryfikuje wzory z podręcznika)

Niech $p = 0.4$, $q = 0.6$, $a = 5$, start $i = 2$. Wtedy $q/p = 1.5$.

$$q_2 = \frac{1.5^5 - 1.5^2}{1.5^5 - 1} = \frac{7.59375 - 2.25}{7.59375 - 1} = \frac{5.34375}{6.59375} \approx 0.8104.$$

Gracz z kapitałem 2 na 5 i niekorzystną grą ($p=0.4$) zbankrutuje z prawd. ok. $81\%$.

Sprawdzimy trzema niezależnymi metodami:
1. wzór jawny (1.11),
2. rozwiązanie układu równań (1.8) macierzowo,
3. symulacja Monte Carlo.

```python
from fractions import Fraction
import numpy as np

p, q, a, start = Fraction(2, 5), Fraction(3, 5), 5, 2

# metoda 1: wzór jawny (1.11) na ułamkach dokładnych
r = q / p                                     # = 3/2
q_i = [(r**a - r**i) / (r**a - 1) for i in range(a + 1)]
print("q_i (wzór 1.11):", [str(x) for x in q_i])
print("q_2 =", q_i[start], "=", float(q_i[start]))
```

```python
# metoda 2: układ równań q_i = p q_{i+1} + q q_{i-1} dla i=1..a-1
# przepisany do postaci A x = b (x = (q_1, ..., q_{a-1}))
A = np.zeros((a - 1, a - 1)); b = np.zeros(a - 1)
for i in range(1, a):
    A[i-1, i-1] = 1.0
    if i - 1 >= 1: A[i-1, i-2] = -float(q)    # wyraz q*q_{i-1}
    else:          b[i-1] += float(q) * 1.0   # q_0 = 1 przechodzi na prawą stronę
    if i + 1 <= a - 1: A[i-1, i] = -float(p)  # wyraz p*q_{i+1}; q_a = 0 nic nie wnosi
x = np.linalg.solve(A, b)
print("q_1..q_4 z układu równań:", np.round(x, 6))
```

```python
# metoda 3: symulacja Monte Carlo ze stałym ziarnem
rng = np.random.default_rng(seed=42)
N = 100_000
ruiny = 0
for _ in range(N):
    kap = start
    while 0 < kap < a:
        kap += 1 if rng.random() < float(p) else -1
    ruiny += (kap == 0)
print(f"symulacja: q_2 ~ {ruiny/N:.4f}   (teoria: {float(q_i[start]):.4f})")

# średni czas gry dla p=q=1/2: D_i = i(a-i)
rng = np.random.default_rng(seed=42)
czasy = []
for _ in range(20_000):
    kap, t = start, 0
    while 0 < kap < a:
        kap += 1 if rng.random() < 0.5 else -1
        t += 1
    czasy.append(t)
print(f"gra sprawiedliwa: E(tau_2) ~ {np.mean(czasy):.2f}   (teoria i(a-i) = {start*(a-start)})")
```

## 📘 Esercizio 1.2 (s. 11 wydruku) — wskazówka

Uogólnienie: cząstka może też **stać w miejscu** z prawd. $r$ (teraz $p + q + r = 1$). Macierz przejścia ma na przekątnej $r$ w stanach wewnętrznych, bariery $0, a$ pochłaniające. Równanie na ruinę:

$$q_i = p\,q_{i+1} + q\,q_{i-1} + r\,q_i \;\Longrightarrow\; (1-r)\,q_i = p\,q_{i+1} + q\,q_{i-1},$$

a ponieważ $1 - r = p + q$, po podzieleniu przez $p+q$ dostajemy **to samo równanie** co przedtem z $p' = p/(p+q)$, $q' = q/(p+q)$. Czekanie w miejscu nie zmienia prawdopodobieństwa ruiny (zmienia tylko czas trwania gry!). Spróbuj wykonać rachunek samodzielnie, zanim porównasz z tą wskazówką.

## Typowe błędy

1. Mylenie $q$ (liczba, $=1-p$) z $q_i$ (prawdopodobieństwo ruiny). Czytaj indeksy!
2. Zgubienie „$+1$" w równaniu na $D_i$ — wtedy wychodzi $D_i \equiv 0$.
3. Rozwiązanie ogólne $q_i = A + B(q/p)^i$ działa tylko dla $p\ne q$; dla $p=q$ pierwiastek jest podwójny i potrzebna baza $\{1, i\}$.
4. Wniosek „gra sprawiedliwa = bezpieczna" jest fałszywy: przy $a\to\infty$ ruina pewna mimo $E(G)=0$.

## Podsumowanie

- Ruina gracza = błądzenie losowe z barierami pochłaniającymi.
- Metoda pierwszego kroku daje równanie różnicowe; rozwiązujemy przez $\alpha^i$ (jak równania liniowe o stałych współczynnikach).
- $q_i = \frac{(q/p)^a - (q/p)^i}{(q/p)^a - 1}$ dla $p\ne q$; $q_i = 1 - i/a$ dla gry sprawiedliwej.
- Średni czas gry: $D_i = i(a-i)$ przy $p = q$.

3. Gracz gra przeciw kasynu ($a \to \infty$) z $p = 0.49$. Czy jego ruina jest pewna?

## Odpowiedzi

1. $q_3 = 1 - 3/10 = 0.7$; $D_3 = 3 \cdot 7 = 21$ partii.
2. Start z $0$ oznacza, że ruina już nastąpiła (prawd. 1), start z $a$ — że gracz już wygrał (ruina niemożliwa). Dla czasu: gra w obu przypadkach już się skończyła, więc trwa 0 partii.
3. Tak: $p = 0.49 < q = 0.51$, a wzór (1.14) mówi, że dla $p \le q$ granica $q_i = 1$ — ruina pewna niezależnie od kapitału początkowego.
